# 01 Sigmoid Activation Function

 # Why Was the Sigmoid Function Introduced?

In the previous section, we learned that a neuron computes a weighted sum:

y = x₁w₁ + x₂w₂ + ... + b

The value of z can be any real number: -100, -8.5, 0, 12, 250

These numbers are difficult to interpret directly.
    
For many machine learning problems, especially binary classification, we want the output to represent a probability.
    
For example:

Toxic or Non-toxic

Active or Inactive

Disease or No Disease

A probability should always lie between 0 and 1.

This is why the Sigmoid activation function was introduced

# What is the Sigmoid Activation Function?

The Sigmoid activation function converts any real number into a value between 0 and 1.

This output can be interpreted as the probability of belonging to a particular class.

## Mathematical Equation

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

Where:

- $\sigma(x)$ = Sigmoid output
- $x$ = Input (weighted sum)
- $e$ = Euler's number (≈ 2.718)

 The Input can be 

-100,
-20,
-5,
0,
3,
20,
100,

 The output will always be between:

0 and 1

# Visualization of Sigmoid function

```python
import numpy as np
import matplotlib.pyplot as plt

# Sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Generate x values
x = np.linspace(-10, 10, 200)
y = sigmoid(x)

# Plot
plt.figure(figsize=(8,5))
plt.plot(x, y, linewidth=2, label='Sigmoid Function')

# Highlight important points
points = [-6, 0, 6]
for p in points:
    plt.scatter(p, sigmoid(p), s=70)
    plt.text(p + 0.3, sigmoid(p), f"({p}, {sigmoid(p):.2f})")

plt.axhline(0.5, linestyle='--')
plt.axvline(0, linestyle='--')

plt.title("Sigmoid Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()
```

# Properties of Sigmoid

Output range: 0 to 1

Smooth and continuous function.
    
Monotonically increasing.
    
Suitable for probability estimation.

# Advantages

✅ Easy to interpret as probabilities.

✅ Works well for binary classification.
    
✅ Smooth and differentiable, which is useful during training.
    
# Limitations
    
❌ Outputs very close to 0 or 1 for large inputs.
    
❌ Can suffer from the vanishing gradient problem, making deep networks harder to train.

❌ Slower convergence than ReLU in deep neural networks.

These limitations are why modern hidden layers often use ReLU, while Sigmoid is commonly reserved for the output layer in binary classification.

# Applications

Sigmoid is commonly used for:

Toxic vs Non-toxic prediction

Active vs Inactive compounds

Disease diagnosis

Binary QSAR classification

ADMET classification tasks

# 02 Tanh ( Hyperbolic Tangent )  Activation Function

# Why Was Tanh Introduced?

The Sigmoid activation function was a major breakthrough, but researchers noticed an important limitation.
The Sigmoid output is always between: 0 and 1

This means all outputs are positive.
Sometimes, having only positive outputs makes learning slower because the activations are not centered around zero.
To address this, the Tanh (Hyperbolic Tangent) activation function was introduced.

# What is the Tanh Activation Function?
The Tanh activation function converts any real number into a value between -1 and +1.
Unlike Sigmoid, Tanh produces both negative and positive outputs, making it zero-centered.

## Mathematical Equation

$$
\tanh(x)=\frac{e^{x}-e^{-x}}{e^{x}+e^{-x}}
$$

Where,

- $e$ = Euler's number (≈ 2.718)
- $x$ = Input to the activation function (Weighted Sum)
- $\tanh(x)$ = Output after activation

```python
import numpy as np
import matplotlib.pyplot as plt

# Tanh activation function
def tanh(x):
    return np.tanh(x)

# Generate x values
x = np.linspace(-10, 10, 200)
y = tanh(x)

# Plot
plt.figure(figsize=(8,5))
plt.plot(x, y, linewidth=2, label='Tanh Function')

# Highlight important points
points = [-6, 0, 6]
for p in points:
    plt.scatter(p, tanh(p), s=70)
    plt.text(p + 0.3, tanh(p), f"({p}, {tanh(p):.2f})")

plt.axhline(0, linestyle='--')
plt.axvline(0, linestyle='--')

plt.title("Tanh Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()
```


x → -∞, Tanh(x) → -1

x = 0, Tanh(x) = 0

x → +∞, Tanh(x) → +1

# Why is Zero-Centered Important?

Imagine a neural network where:

Some neurons should increase the prediction.

Some neurons should decrease the prediction.

With Tanh:

Positive values support one direction.

Negative values support the opposite direction.

This balance often helps optimization during training

# Advantages

✅ Zero-centered output.

✅ Often converges faster than Sigmoid.
    
✅ Smooth and differentiable.

# Limitations

❌ Still suffers from the vanishing gradient problem.
    
❌ Large positive and negative inputs saturate near +1 and -1, causing gradients to become very small.

# 03_ReLU ( Rectified Linear Unit ) 

# Why Was ReLU Introduced?

We learned that both Sigmoid and Tanh have a major limitation.

They suffer from the Vanishing Gradient Problem.

When the input becomes very large or very small,

their output becomes almost constant.

Sigmoid

Large Positive → 1

Large Negative → 0

Tanh

Large Positive → +1

Large Negative → -1

As a result,

the gradients become extremely small,
making deep neural networks learn very slowly.

Researchers wanted an activation function that
is computationally simple,
trains faster,
avoids the vanishing gradient problem (for positive inputs).

This led to the ReLU activation function.

# What is ReLU?

ReLU (Rectified Linear Unit) outputs the input directly if it is positive; otherwise, it outputs zero.

Unlike Sigmoid and Tanh,
ReLU does not compress positive values into a limited range.

## Mathematical Equation

$$
\text{ReLU}(x)=
\begin{cases}
0, & x<0 \\
x, & x\ge0
\end{cases}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# ReLU function
def relu(x):
    return np.maximum(0, x)

x = np.linspace(-10,10,200)
y = relu(x)

plt.figure(figsize=(8,5))
plt.plot(x,y,linewidth=2,label="ReLU")

points=[-6,0,6]

for p in points:
    plt.scatter(p,relu(p),s=70)
    plt.text(p+0.3,relu(p),f"({p},{relu(p):.1f})")

plt.axhline(0,linestyle="--")
plt.axvline(0,linestyle="--")

plt.title("ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()
```

 # Example 

Suppose a neuron predicts molecular toxicity.

Weighted Sum : z = -4

ReLU : Output = 0

Another molecule z = 6

ReLU : Output = 6

Instead of compressing positive values,
ReLU keeps them unchanged,
allowing the network to learn faster

Advantages

✅ Very simple mathematical function.

✅ Fast computation.

✅ Reduces the vanishing gradient problem for positive inputs.

✅ Faster convergence.

✅ Most widely used activation function in deep learning.

Limitations

❌ Neurons receiving only negative inputs always output 0.

❌ These neurons may stop learning, a problem known as the Dying ReLU Problem.

We'll solve this in the next topic using Leaky ReLU.

# Applications

ReLU is widely used in:

CNN
Deep Neural Networks

Graph Neural Networks (GNNs)

Autoencoders

Computer Vision

Molecular Property Prediction

QSAR Models

ADMET Prediction

# 04_Leaky ReLU Activation Function

Why Was Leaky ReLU Introduced?

In the previous topic, we learned that ReLU works as:

If x < 0  →  Output = 0

If x ≥ 0  →  Output = x

This makes ReLU very fast and efficient.
However, it has one major limitation.

The Dying ReLU Problem

Suppose a neuron receives negative inputs repeatedly.
Example:

-5

-8

-2

-10

Using ReLU:

ReLU(-5)  = 0

ReLU(-8)  = 0

ReLU(-2)  = 0

ReLU(-10) = 0

The neuron always outputs: 0

Eventually,
the neuron stops learning because it never becomes active.
This is called the Dying ReLU Problem.

# What is Leaky ReLU?

Leaky ReLU is a modified version of ReLU.

Instead of making all negative values equal to zero,

it allows a small amount of negative information to pass through.
Therefore,
the neuron can continue learning even for negative inputs.

## Mathematical Equation

$$
\text{Leaky ReLU}(x)=
\begin{cases}
0.01x, & x<0 \\
x, & x\ge0
\end{cases}
$$

Here,
0.01 is called the leak coefficient (or negative slope).

Some implementations use values like 0.1 or 0.2, but 0.01 is the most common example.

Understanding the Equation

If x = -10

Output

0.01 × (-10) =  -0.1

If  x = -5

Output: -0.05

If  x = 4

Output: 4
Notice that negative inputs are not completely blocked.

```python
import numpy as np
import matplotlib.pyplot as plt

def leaky_relu(x):
    return np.where(x > 0, x, 0.01*x)

x = np.linspace(-10,10,200)
y = leaky_relu(x)

plt.figure(figsize=(8,5))
plt.plot(x,y,linewidth=2,label="Leaky ReLU")

points=[-6,0,6]

for p in points:
    plt.scatter(p,leaky_relu(p),s=70)
    plt.text(p+0.3,leaky_relu(p),f"({p},{leaky_relu(p):.2f})")

plt.axhline(0,linestyle="--")
plt.axvline(0,linestyle="--")

plt.title("Leaky ReLU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("Output")
plt.grid(True)
plt.legend()

plt.show()
```


Observe the graph carefully:

Positive side is identical to ReLU.

Negative side has a small slope instead of being completely flat.

This small slope allows gradients to flow during training.

# Advantages

✅ Solves the Dying ReLU Problem.

✅ Allows gradients to flow for negative inputs.

✅ Faster convergence.

✅ Often performs better than standard ReLU.
    
# Limitations
    
❌ The leak coefficient (0.01) is fixed.

❌ It may not always be the optimal value for every dataset.
    
To address this, researchers developed Parametric ReLU (PReLU), where the negative slope is learned during training.

# Applications
    
Leaky ReLU is commonly used in:

Deep Neural Networks

CNNs

Graph Neural Networks

Molecular Property Prediction

ADMET Prediction
Toxicity Prediction

# 05 Softmax Activation Function



# Why was Softmax introduced?
Suppose we want to classify a molecule into one of three classes:

Acid

Base

Neutral

These are called logits.

The problem is that these numbers aren't probabilities:
They don't lie between 0 and 1.
They don't add up to 1.
We need a function that converts these raw scores into a probability distribution.

That's where Softmax comes in. Generally used for multi-class classification

# What is Softmax?

Softmax converts a set of raw scores (logits) into probabilities that are between 0 and 1 and sum to 1.

For example:
Raw scores

Acid     → 2.5

Base     → 1.2

Neutral  → 0.3

          ↓ Softmax

Probabilities

Acid     → 0.71

Base     → 0.21

Neutral  → 0.08

Total    → 1.00

So the model would predict:
Acid, with approximately 71% probability.

## Mathematical Equation

$$
\mathrm{Softmax}(z_i)=
\frac{e^{z_i}}
{\sum_{j=1}^{n}e^{z_j}}
$$

Where:

(z_i) = raw score (logit) for class (i)

(e) = Euler's number

(n) = total number of classes

denominator = sum of exponentials of all class scores

Softmax output = probability for that class


Why is there an (e^z)?

This is important.

The exponential function makes differences between the scores more pronounced.
    
For example:

Score A = 3

Score B = 1

After exponentiation:

e³ ≈ 20.09

e¹ ≈ 2.72

So the higher score gets a much larger contribution.
    
Then Softmax divides each exponential value by the total, converting them into probabilities.

# Simple Example

Suppose our model produces:

Acid     = 2

Base     = 1

Neutral  = 0

First calculate exponentials:

e² ≈ 7.39

e¹ ≈ 2.72

e⁰ = 1

Total:

7.39 + 2.72 + 1 = 11.11

Now:
Acid
7.39 / 11.11 ≈ 0.665

Base
2.72 / 11.11 ≈ 0.245

Neutral
1 / 11.11 ≈ 0.090


```python
# Visualization

import numpy as np
import matplotlib.pyplot as plt

# Softmax function
def softmax(z):
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z)

# Raw model outputs (logits)
classes = ["Acid", "Base", "Neutral"]
logits = np.array([2, 1, 0])

# Apply Softmax
probabilities = softmax(logits)

# Display probabilities
for cls, prob in zip(classes, probabilities):
    print(f"{cls}: {prob:.3f}")

# Plot probabilities
plt.figure(figsize=(8, 5))
plt.bar(classes, probabilities)

plt.title("Softmax Activation Function")
plt.xlabel("Class")
plt.ylabel("Probability")
plt.ylim(0, 1)

plt.grid(axis="y")
plt.show()
```

# Where is Softmax Used?

Softmax is commonly used in the output layer for multi-class classification.
Examples:

Molecular class prediction

Protein classification

Compound activity classification

Multi-class toxicity classification

Cell-type classification

Image classification

# Softmax Limitation

Softmax assumes the classes are mutually exclusive.


For example:

Acid

Base

Neutral

A molecule is expected to belong to one class.

But suppose you have:

CYP3A4 inhibitor

CYP2D6 inhibitor

CYP2C9 inhibitor

A compound can potentially inhibit multiple enzymes simultaneously.

That is a multi-label classification problem.

In such a case, we would generally use separate Sigmoid outputs, rather than Softmax.

This distinction is very important in your ADMET work.

# 06 Exponential Linear Unit  ( ELU )

ELU = Exponential Linear Unit

It behaves like ReLU for positive values, but uses an exponential function for negative values.

## Mathematical Equation

$$
\mathrm{ELU}(x)=
\begin{cases}
x, & x>0\\
\alpha(e^x-1), & x\leq0
\end{cases}
$$

Where $\alpha$ controls the negative saturation value. 

```python
import numpy as np
import matplotlib.pyplot as plt

# ELU activation function
def elu(x, alpha=1.0):
    return np.where(x > 0, x, alpha * (np.exp(x) - 1))

# Generate x values
x = np.linspace(-10, 10, 200)

# Calculate ELU values
y = elu(x)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, linewidth=2, label="ELU")

plt.axhline(y=0, linestyle="--")
plt.axvline(x=0, linestyle="--")

plt.title("ELU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("ELU(x)")
plt.xlim(-10, 10)
plt.ylim(-2, 10)

plt.grid(True)
plt.legend()

plt.show()
```

# 07_ GELU ( Gaussian Error Linear Unit )

# Why was GELU introduced?

We've already seen ReLU:

ReLU makes a very sharp decision:

x < 0  →  0

x > 0  →  x

GELU takes a different approach.

Instead of completely removing negative values, it smoothly controls how much of the input should pass through.

So the basic idea is:

GELU provides a smooth alternative to ReLU


#  What is GELU?

GELU = Gaussian Error Linear Unit

It is a smooth, nonlinear activation function that weights the input according to its magnitude rather than using a hard cutoff like ReLU.
    
In simple terms:
    
ReLU:
    
"Negative? → Remove it."

GELU:

"Negative or positive? → Decide smoothly how much to pass."

## Mathematical Equation

$$
\mathrm{GELU}(x)
=
x\Phi(x)
$$

where $\Phi(x)$ is the cumulative distribution function of the standard normal distribution.

$$
\mathrm{GELU}(x)
\approx
\frac{1}{2}x
\left[
1+\tanh
\left(
\sqrt{\frac{2}{\pi}}
\left(x+0.044715x^3\right)
\right)
\right]
$$

GELU doesn't make an abrupt jump at zero.

It produces a smooth transition:

Negative → gradually approaches negative values

Near 0   → smoothly transitions

Positive → increasingly behaves like x

That's the key idea.

# Visualization

```python
import numpy as np
import matplotlib.pyplot as plt

# GELU activation function
def gelu(x):
    return 0.5 * x * (
        1 + np.tanh(
            np.sqrt(2 / np.pi) * (x + 0.044715 * x**3)
        )
    )

# Generate x values
x = np.linspace(-5, 5, 300)

# Calculate GELU
y = gelu(x)

# Plot
plt.figure(figsize=(8, 5))
plt.plot(x, y, linewidth=2, label="GELU")

plt.axhline(y=0, linestyle="--")
plt.axvline(x=0, linestyle="--")

plt.title("GELU Activation Function")
plt.xlabel("Input (x)")
plt.ylabel("GELU(x)")
plt.grid(True)
plt.legend()

plt.show()
```
