# 19. Attention Mechanism

## 1. What is Attention?

Attention is a mechanism that allows a neural network to selectively focus on the most relevant parts of an input sequence when producing an output.

In simple words:

> Attention allows the model to decide which parts of the input are important for the current task.

Instead of treating every input element equally, the model assigns different importance to different elements.

# 14. Scaled Dot-Product Attention

The most important attention equation used in Transformers is:

$$
\boxed{
Attention(Q,K,V)
=
softmax
\left(
\frac{QK^T}{\sqrt{d_k}}
\right)V
}
$$

Where:

- $Q$ = Query matrix
- $K$ = Key matrix
- $V$ = Value matrix
- $d_k$ = Dimension of the key vectors

This is called Scaled Dot-Product Attention.

---

# 15. Why Do We Divide by √dₖ?

When the dimensionality of the key vectors becomes large, dot products can become large in magnitude.

Large values passed into softmax can make the distribution extremely sharp.

This can lead to very small gradients.

Therefore, the score is scaled:

$$
\frac{QK^T}{\sqrt{d_k}}
$$

This helps keep the values in a more suitable range.

---

# 39. Attention and Computational Cost

Standard self-attention creates an interaction between many pairs of sequence positions.

For sequence length:

$$
n
$$

the attention matrix has approximately:

$$
n\times n
$$

entries.

Therefore, standard self-attention has quadratic scaling with sequence length in its core attention computation.

Conceptually:


This is one of the important considerations when designing Transformer systems.
