# 📓 9. Epoch, Batch & Iteration

## 1. What is an Epoch?

An epoch means that the neural network has gone through the entire training dataset once.

For example, suppose we have:

$$
10,000
$$

training molecules.

If the model processes all 10,000 molecules once, that is:

$$
\boxed{1\text{ Epoch}}
$$

If it processes the entire dataset again:

$$
\boxed{2\text{ Epochs}}
$$

And so on.

---

## Why Do We Need Multiple Epochs?

One pass through the dataset is usually not enough for the model to learn the underlying patterns.

The model repeatedly sees the training data and updates its parameters

Ideally, the loss decreases as training progresses

---

## Simple Example

Suppose our training dataset contains:

$$
1,000
$$

molecules.

If we train for:

$$
10\text{ epochs}
$$

the model will go through the training dataset 10 times.

Conceptually:


This does not mean that the model simply memorizes the dataset. During training, the weights are updated as the model processes the data.

---

## Epoch vs Training Step

An epoch means:

> The complete training dataset has been processed once.

A training step refers to a parameter-update step.

When we later introduce batches, one epoch can contain many training steps.

For example:

$$
10,000\text{ samples}
$$

with a batch size of:

$$
100
$$

gives:

$$
\frac{10,000}{100}=100
$$

training steps per epoch.

Therefore:

$$
\boxed{1\text{ epoch}=100\text{ training steps}}
$$

in this example.

# 2. What is a Batch?

A batch is a smaller group of training samples taken from the complete training dataset.

Instead of giving the entire dataset to the neural network at once, we can divide it into smaller groups called batches.

For example, suppose we have:

$$
10,000
$$

training molecules.

If:

$$
Batch\ Size=100
$$

then the dataset is divided approximately as:

So there are:

$$
\frac{10,000}{100}=100
$$

batches.

---

# Why Do We Use Batches?

Imagine trying to process millions of molecules simultaneously.

That can require a very large amount of memory.

Instead, we can process the data in smaller groups.

This makes training more practical and efficient.

---

# What Happens With One Batch?

For each batch, the model performs the normal learning process:


Then it moves to the next batch.

---

# Batch Size

The number of samples contained in one batch is called the batch size.

For example:

batch_size = 32

means the model processes approximately 32 training samples at a time.

Common batch sizes include:


The appropriate value depends on the dataset, model, available memory, and training setup.

# Batch vs Entire Dataset

There are three important approaches:

### Full Dataset

The entire training dataset is used for one parameter update.

This is called:

Batch Gradient Descent

---

### One Sample

One training sample is used for each parameter update.

This is called:

Stochastic Gradient Descent (SGD)

---

### Small Groups

A small group of samples is used for each update.

This is called:

Mini-Batch Gradient Descent

This is the approach commonly used when training neural networks.

# 3. What is an Iteration?

An iteration is one complete parameter-update step during training.

# General Formula

If the dataset contains:

$$
N
$$

samples and the batch size is:

$$
B
$$

then the approximate number of iterations per epoch is:

$$
\boxed{
\text{Iterations per Epoch}
=
\frac{N}{B}
}
$$

If the division is not exact, the final smaller batch is normally still processed, so in practice we often use:

$$
\boxed{
\text{Iterations per Epoch}
=
\left\lceil\frac{N}{B}\right\rceil
}
$$

where $\lceil\rceil$ means rounding up.

---

# 4. Types of Gradient Descent

There are three common ways to decide how much training data should be used before updating the model's parameters:

1. Batch Gradient Descent
2. Stochastic Gradient Descent (SGD)
3. Mini-Batch Gradient Descent

The main difference is:

> How many training samples are used to calculate the gradient before updating the weights?

---

# 1. Batch Gradient Descent

In Batch Gradient Descent, the model uses the entire training dataset to calculate the gradient before making one parameter update.

Suppose we have:

$$
10,000
$$

training samples.

The model processes all 10,000 samples:

So:

$$
\boxed{1\text{ update per epoch}}
$$

### Advantage

The gradient is calculated using the entire dataset, so the update is generally stable.

### Disadvantage

For a very large dataset, this can require a lot of memory and computation before each update.

---

# 2. Stochastic Gradient Descent (SGD)

In Stochastic Gradient Descent, the model uses one training sample at a time to calculate the gradient and update the parameters.

For example:


If we have 10,000 samples:

$$
\boxed{10,000\text{ updates per epoch}}
$$

### Advantage

The model updates its parameters very frequently.

### Disadvantage

Because each update is based on only one sample, the updates can be noisy.

The model may move toward lower loss but with more fluctuations.

---

# 3. Mini-Batch Gradient Descent

Mini-Batch Gradient Descent uses a small group of samples to calculate the gradient before updating the parameters.

For example:

$$
Batch\ Size=32
$$

If we have:

$$
10,000
$$

samples and:

$$
Batch\ Size=100
$$

then approximately:

$$
\frac{10,000}{100}=100
$$

updates occur per epoch.

---


# Comparison

| Method | Samples per update | Updates per epoch | Main characteristic |
|---|---:|---:|---|
| Batch Gradient Descent | Entire dataset | 1 | Stable but computationally heavy |
| SGD | 1 sample | $N$ | Frequent but noisy updates |
| Mini-Batch | Small group | Approximately $N/B$ | Practical balance |

Where:

- $N$ = Number of training samples
- $B$ = Batch size
