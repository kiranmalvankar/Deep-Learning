# What is Forward Propagation?

Simple Definition

Forward propagation is the process of passing input data through a neural network, layer by layer, until the network produces its final prediction.

In other words:

Input

  ↓
  
Hidden Layer 1

  ↓
  
Hidden Layer 2

  ↓
  
...

  ↓
  
Output Layer

  ↓
  
Prediction

The information always moves forward.
That's why it's called forward propagation.

# Forward propagation is the process of passing input data through the layers of a neural network, where each neuron calculates a weighted sum, adds a bias, applies an activation function, and passes the resulting output to the next layer until the final prediction is produced.

## Mathematical Equations of Forward Propagation

For a neuron, the weighted sum is:

$$
z = \sum_{i=1}^{n} x_iw_i + b
$$

The activation function is then applied:

$$
a = f(z)
$$

Therefore, the complete operation is:

$$
a = f\left(\sum_{i=1}^{n}x_iw_i+b\right)
$$

Where:

- $x_i$ = Input
- $w_i$ = Weight
- $b$ = Bias
- $z$ = Weighted sum
- $f$ = Activation function
- $a$ = Activated output

# Forward Propagation through hidden layers

A hidden layer is a collection of neurons. Each neuron receives the same previous-layer outputs but uses its own weights and bias, producing different transformed outputs.

# Matrix Representation of Forward Propagation \

Why do we need matrices?

Earlier, we calculated each neuron separately.

For example:

Neuron 1:

z₁ = x₁w₁ + x₂w₂ + b₁

Neuron 2:

z₂ = x₁w₃ + x₂w₄ + b₂

That works for 2 neurons.

But imagine a real neural network with:

100 input features

        ×
        
256 neurons

Writing hundreds of individual equations would be impractical.
|
So neural networks use matrix multiplication to calculate an entire layer at once.
