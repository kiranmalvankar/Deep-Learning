# 1. Overfitting and Underfitting

## Why Do We Need to Understand This?

During training, our goal is not simply to make the model perform well on the training data.

We want the model to learn patterns that can also work on unseen data.

For example, in cheminformatics:

A model can fail in two different ways:

1. Underfitting
2. Overfitting

---

# 1. Underfitting

Underfitting happens when the model is too simple to learn the important patterns in the data.

For example:

The model performs poorly on:

- Training data
- Unseen/test data

Conceptually:

- Training performance → Poor
- Test performance     → Poor


### Example

Suppose the actual relationship between molecular descriptors and solubility is complex.

If we use a model that is too simple, it may fail to capture the relationship.

The model has not learned enough.

---

# 2. Overfitting

Overfitting happens when the model learns the training data too closely, including patterns that do not generalize to new data.

The model performs very well on the training data but poorly on unseen data.

- Training performance → Very good
- Test performance     → Poor

# 2. Bias vs Variance

## What is Bias?

Bias represents the error caused when a model is too simple and makes assumptions that are too strong about the relationship in the data.

A high-bias model tends to underfit the data.

### Example

Suppose the actual relationship between molecular features and a property is highly non-linear.

If we use an overly simple linear model:

the model may not be able to capture the important molecular relationships.

This results in:

$$
\boxed{\text{High Bias}}
$$

and usually:

$$
\boxed{\text{Underfitting}}
$$

---

# What is Variance?

Variance represents how much the model's learned predictions change when the training data changes.

A high-variance model is very sensitive to the particular training data it sees.

It can learn the training data too specifically.


This results in:

$$
\boxed{\text{High Variance}}
$$

and usually:

$$
\boxed{\text{Overfitting}}
$$

---

# Bias vs Variance

The basic relationship is:

| Situation | Bias | Variance | Typical Problem |
|---|---|---|---|
| Underfitting | High | Low | Model too simple |
| Good fit | Appropriate | Appropriate | Good generalization |
| Overfitting | Low | High | Model too complex |

---

# Simple Analogy

Imagine trying to draw a curve through experimental data.

### High Bias

The model is too rigid:

It cannot follow the underlying pattern.

---

### High Variance

The model tries to follow every individual observation:

It may also be learning random noise.

---

### Good Balance

The model captures the underlying pattern without following every small fluctuation:

---

# Bias-Variance Tradeoff

In machine learning, we generally want to find a balance between bias and variance.

Increasing model complexity can reduce bias but may increase variance.

Reducing model complexity can reduce variance but may increase bias.

Therefore, the goal is not simply:

> "Make the model as complex as possible."

The goal is:

> Find a model that captures the important patterns while generalizing to unseen data.

---



# Important Point

Bias and variance are not exactly the same thing as training and test error.

They are concepts that help us understand why a model's error behaves the way it does.

A simplified view is:

$$
\boxed{
\text{High Bias} \rightarrow \text{Underfitting}
}
$$

$$
\boxed{
\text{High Variance} \rightarrow \text{Overfitting}
}
$$

---

# Key Takeaway

> Bias is associated with models that are too simple, while variance is associated with models that are too sensitive to the training data.

# 3. Train, Validation, and Test Sets

## Why Do We Split the Dataset?

Suppose we have a dataset of known molecules.

We want to build a model that can predict a property for unseen molecules.

If we use every molecule for training, we don't have an independent dataset to properly evaluate how the model performs on unseen data.

Therefore, we divide the dataset into different parts.

When model selection and tuning are involved, we usually use three sets:

Complete Dataset
       ↓
 ┌─────┬──────────┬
 ↓     ↓          ↓
Train  Validation Test


---

# 1. Training Set

The training set is the data used to actually train the model.

The model uses these molecules to learn:

- Weights
- Biases
- Relationships between features and target

For example:

The model is allowed to learn from the training data.

---

# 2. Validation Set

The validation set is used during model development to evaluate different model choices and settings.

For example, we may want to decide:

- Which model performs better?
- What hyperparameters should we use?
- How many layers should we use?
- What learning rate should we use?
- When should training stop?

The validation data helps us make these decisions without using the final test set.

---

# 3. Test Set

The test set is kept separate until the final evaluation.

The model should not use the test molecules to learn its parameters or tune its hyperparameters.

After we have finalized the model:

Final Model
     ↓
Test Set
     ↓
Final Evaluation


This gives us a better estimate of how the model may perform on genuinely unseen data.

# 5. Detecting Overfitting

## Why Do We Need to Detect Overfitting?

During training, the model's performance on the training data usually improves.

But this does not guarantee that the model is becoming better at predicting unseen data.

We therefore monitor both:

- Training performance
- Validation performance

This helps us identify whether the model is learning useful patterns or simply becoming too specialized to the training data.

---

# Training Loss vs Validation Loss

A common way to detect overfitting is to monitor the loss during training.

Later:

Generalization is the ability of a machine-learning model to perform well on new, unseen data.

This is one of the most important goals of machine learning.

A model should not simply memorize the training molecules.

It should learn meaningful patterns that can be applied to molecules it has never seen before.

There is an important distinction between:

### Unseen test molecules

Molecules that were held out from the dataset during model development.
