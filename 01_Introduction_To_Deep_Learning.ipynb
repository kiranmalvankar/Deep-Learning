# What is Deep Learning?

Deep Learning (DL) is a branch of Machine Learning (ML) that uses Artificial Neural Networks (ANNs) with multiple hidden layers to automatically learn complex patterns from data.

Unlike traditional machine learning, where features are often manually designed, deep learning learns the most useful features directly from the data.

In Deep Learning

> Instead of relying only on manually designed features, a neural network learns complex relationships directly from the data through multiple layers.
> 
> For molecular problems, it can learn representations from:
- Molecular descriptors
- Fingerprints
- SMILES sequences
- Molecular graphs (using GNNs)
This enables the model to capture patterns that may be difficult to engineer manually.

# Biological Neuron VS Artificial Neuron

# What is a Biological Neuron?

A biological neuron is the basic functional unit of the nervous system. It receives information, processes it, and transmits signals to other neurons.
The human brain contains approximately 86 billion neurons, connected through trillions of synapses.

These interconnected neurons enable us to:

Learn

Remember

Recognize images

Understand language

Make decisions

# Structure of a Biological Neuron

A biological neuron consists of four main parts:

# 1. Dendrites

Receive signals from other neurons.

Think of them as the input channels.

Example: Your eyes see a molecule, and the information enters the brain through neurons.

# 2. Cell Body (Soma)

Processes all incoming signals.

Decides whether enough information has been received to generate an output signal.

This is the neuron's decision-making center.

# 3. Axon

Carries the electrical signal away from the cell body.

Transmits the output to other neurons.

# 4. Synapse

The connection between two neurons.

Signals pass across the synapse to the next neuron.

This allows information to travel through the neural network.

# What is an Artificial Neuron?

An Artificial Neuron is a mathematical model inspired by a biological neuron. It receives multiple input values, processes them using mathematical operations, and produces an output.

Just like a biological neuron:

It receives information.

It processes the information.

It passes the result to the next neuron.

Scientists observed that the human brain learns by connecting billions of neurons. They wanted computers to learn in a similar way.
To achieve this, they designed a simplified mathematical model called an Artificial Neuron.

An artificial neuron is the basic building block of all neural networks. Every deep learning model—ANN, CNN, RNN, Transformers, and GNNs—is built by connecting many artificial neurons together

Structure of an Artificial Neuron
An artificial neuron consists of four main components.

# 1. Inputs (x₁, x₂, x₃, …)

These are the values given to the neuron.

For example, in a drug discovery project, the inputs could be:

Molecular Weight

LogP

TPSA

Number of Hydrogen Bond Donors

Number of Hydrogen Bond Acceptors

These features are represented as:
x₁, x₂, x₃, x₄, x₅

The inputs are similar to the dendrites of a biological neuron because they receive information.

# 2. Weights (w₁, w₂, w₃, …)

Each input has a corresponding weight.

Weights determine how important each input is.

For example:

If LogP is more important than Molecular Weight for predicting permeability, the neuron will assign a larger weight to LogP.
The network learns these weights automatically during training.

# 3. Processing Unit

The neuron combines all the weighted inputs and adds a bias.

Mathematically:  Output = (x₁ × w₁) + (x₂ × w₂) + ... + Bias

This is the neuron's decision-making step, similar to the cell body (soma) in a biological neuron.


# 4. Activation Function

After processing, the result passes through an activation function.

The activation function decides whether the neuron should produce a strong output, a weak output, or no output at all.

This is similar to how a biological neuron decides whether to transmit a signal.

We'll study activation functions in detail later.

# 1. What are Inputs?

Inputs are the features or information provided to an artificial neuron.
In machine learning, inputs are usually called features (X).

These are represented as:x₁, x₂, x₃, x₄, x₅
These values enter the neuron.

# 2. What are Weights?

Now think like a scientist.

Will every feature contribute equally?

No.

For permeability prediction:
LogP might be very important.
Molecular Weight might be moderately important.

Another descriptor might have little influence.

The neuron represents this importance using weights.

Each input gets its own weight

x₁ → w₁

x₂ → w₂

x₃ → w₃

A larger weight means the feature has a greater influence on the prediction.
A smaller weight means the feature contributes less.

The amazing part is that the model learns these weights automatically during training.
You don't assign them manually.


# 3. What is Bias?

This is the concept that confuses most beginners.

Think of bias as a starting point or an adjustment value.

Imagine a weighing scale.

Sometimes, even when nothing is placed on it, it shows +0.2 kg.

You calibrate it to make the measurements accurate.

The bias plays a similar role.

It helps the neuron adjust its output and learn more flexible decision boundaries.
Without bias, the neuron would be too restricted in the patterns it can learn.

# "Why isn't the weighted sum itself the final answer? Why do we need an activation function?"

# What is an Activation Function?

An activation function is a mathematical function that takes the weighted sum (y) as input and produces the neuron's final output.
It determines whether the neuron should activate and how strongly it should respond

# Why need the Activation Function ?

Imagine building a neural network with:

10 layers

100 neurons per layer

If every neuron only performs: Weighted Sum

then the entire network is still just a linear model.
    
No matter how many layers you add, it cannot learn complex relationships.
    
The activation function introduces non-linearity, allowing neural networks to solve complex problems such as:

Image recognition
Speech recognition

Molecular property prediction

Protein structure prediction

# Why is Non-Linearity Important?

Many real-world relationships are not linear.

For example:

Increasing LogP does not always increase permeability.

Sometimes:

Too low → Poor permeability.

Moderate → Good permeability.

Too high → Poor permeability again.

This is a non-linear relationship.

Deep learning models can learn these patterns because of activation functions.
