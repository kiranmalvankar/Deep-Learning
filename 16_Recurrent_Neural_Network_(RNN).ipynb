## 1. What is an RNN?

Recurrent Neural Network (RNN) is a type of neural network designed to process sequential or time-dependent data.

Unlike a basic ANN, an RNN maintains information from previous time steps and uses it when processing the current input.

In simple words:

> RNN has a form of memory that allows information from previous steps to influence the current prediction.

# 5. The Core Idea of RNN

At each time step, an RNN receives:

Mathematically:

$$
\boxed{
h_t=f(W_xx_t+W_hh_{t-1}+b)
}
$$

Where:

- $x_t$ = Current input
- $h_{t-1}$ = Previous hidden state
- $h_t$ = Current hidden state
- $W_x$ = Input-to-hidden weights
- $W_h$ = Hidden-to-hidden weights
- $b$ = Bias
- $f$ = Activation function

---

# 24. Backpropagation Through Time (BPTT)

Because RNNs process sequences over multiple time steps, backpropagation is performed through the unrolled sequence.

This is called:

Backpropagation Through Time (BPTT).

Conceptually:

Gradients flow backward through the sequence.

---

# 25. Why Is BPTT Needed?

The current hidden state depends on previous hidden states.

Therefore, a parameter used at one time step also influences later outputs.

BPTT allows the model to calculate how the loss is affected by parameters across the sequence.

---

# 26. The Vanishing Gradient Problem

One major limitation of traditional RNNs is the vanishing gradient problem.

During BPTT, gradients can become extremely small as they are propagated backward through many time steps.

Conceptually:

Eventually, the gradient may become so small that earlier time steps learn very slowly.

---

# 27. Why Is Vanishing Gradient a Problem?

Suppose the sequence is:

x₁ → x₂ → x₃ → ... → x₁₀₀


The model may need information from $x_1$ when making a prediction at $x_{100}$.

But if the gradient becomes very small while propagating backward:

the network may struggle to learn long-term dependencies.

---

# 28. Exploding Gradient Problem

The opposite can also happen.

Gradients can become extremely large:

This is called the exploding gradient problem.

It can cause unstable training.

# 31. RNN vs CNN

| Feature | RNN | CNN |
|---|---|---|
| Designed for | Sequential data | Spatial/grid-like data |
| Main mechanism | Recurrent connections | Convolution |
| Maintains hidden state | Yes | No recurrent hidden state |
| Parameter sharing | Across time steps | Across spatial locations |
| Typical applications | Sequence/time series | Images/spatial data |
| Molecular representation | SMILES sequences | Molecular images/grids |

---

# 32. RNN vs ANN

| Feature | ANN | RNN |
|---|---|---|
| Sequential information | Not inherently modeled | Yes |
| Hidden state | No recurrent hidden state | Yes |
| Input order | Usually not explicitly modeled | Important |
| Memory across steps | No | Yes |
| Typical use | Tabular/general features | Sequence/time-series data |
