# Loss Function vs Evaluation Metric

Both loss functions and evaluation metrics are used to measure model performance, but they serve different purposes.

---

## 1. Loss Function

A loss function measures how far the model's predictions are from the actual values during training.

The loss is used by the learning process to improve the model.

The general training process is:

$$
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Gradient}
\rightarrow
\text{Weight Update}
$$

The objective of training is usually to minimize the loss.

### Examples

For regression:

- MAE
- MSE

For binary classification:

- Binary Cross-Entropy

For multi-class classification:

- Categorical Cross-Entropy

---

## 2. Evaluation Metric

An evaluation metric is used to understand how well the trained model performs.

Metrics help us assess and compare models using measures that are meaningful for the particular problem.

Examples include:

### Regression

- MAE
- MSE
- RMSE
- $R^2$

### Classification

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC

---

## 3. Main Difference

The simplest way to understand the difference is:

> Loss function → Helps the model learn.

> Evaluation metric → Helps us evaluate the model.

---

## 4. Example: Regression

Suppose we are predicting molecular permeability.

The model produces:

$$
\text{Prediction}
$$

We calculate a loss:

$$
MSE
$$

Then:

$$
MSE
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Gradient}
\rightarrow
\text{Weight Update}
$$

The model repeats this process during training to reduce the loss.

After training, we may evaluate the model using:

$$
MAE,\quad RMSE,\quad R^2
$$

These metrics help us understand how well the model performs on unseen data.

---

## 5. Why Can the Same Metric Be Both?

Some measures can be used as both a loss function and an evaluation metric.

For example:

$$
MAE
$$

can be used to:

- Measure the error during training
- Evaluate the final model

Similarly:

$$
MSE
$$

can also serve both purposes.

However, some evaluation metrics are not commonly used directly as training losses.

For example:

Accuracy is useful for evaluating classification performance, but it is generally not used directly as the optimization loss for a neural network.

---

## 6. Training vs Evaluation

### During Training

The model needs a differentiable signal that can guide weight updates.

$$
\text{Input}
\rightarrow
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Weight Update}
$$

### During Evaluation

We want to answer questions such as:

- How accurate is the model?
- How large are the prediction errors?
- Does the model generalize to unseen data?
- Is the model better than another model?

Therefore, we calculate appropriate evaluation metrics.

---

## Simple Comparison

| Loss Function | Evaluation Metric |
|---|---|
| Used during model training | Used to assess model performance |
| Guides weight optimization | Helps us understand model quality |
| Usually needs to provide a useful gradient | Does not necessarily need to be differentiable |
| Objective is generally to minimize it | Used for comparison and interpretation |

---

## Key Takeaway

> A loss function provides the learning signal used to optimize the model, while an evaluation metric tells us how well the model is performing.

They can sometimes be the same mathematical quantity, but their purpose in the machine-learning workflow is different.

# Model Evaluation

## What is Model Evaluation?

Model evaluation is the process of measuring how well a trained machine learning or deep learning model performs on data.

The purpose is to determine whether the model:

- Makes accurate predictions
- Generalizes to unseen data
- Performs consistently
- Is suitable for the intended application

A model should not be judged only by how well it performs on the training data.

---

## Why Do We Need Model Evaluation?

During training, a model learns patterns from the training data.

However, our real goal is to make good predictions on new, unseen data.

Therefore, we need to evaluate the model using data that was not used to train it.

The general workflow is:

$$
\text{Training Data}
\rightarrow
\text{Model Training}
\rightarrow
\text{Trained Model}
$$

Then:

$$
\text{Unseen Data}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Predictions}
\rightarrow
\text{Evaluation Metrics}
$$

---

## Training vs Evaluation

### During Training

The model learns its parameters by minimizing a loss function.

$$
\text{Input}
\rightarrow
\text{Prediction}
\rightarrow
\text{Loss}
\rightarrow
\text{Backpropagation}
\rightarrow
\text{Weight Update}
$$

### During Evaluation

We use evaluation metrics to measure the quality of the model's predictions.

$$
\text{Unseen Data}
\rightarrow
\text{Prediction}
\rightarrow
\text{Evaluation Metrics}
$$

---

## What Are Evaluation Metrics?

Evaluation metrics are numerical measures used to quantify model performance.

Different problems require different evaluation metrics.

### Regression

When the target is a continuous value, commonly used metrics include:

- MAE
- MSE
- RMSE
- $R^2$

### Classification

When the target represents classes, commonly used metrics include:

- Accuracy
- Precision
- Recall
- F1-score
- ROC-AUC

---

## Key Idea

> Model evaluation tells us how well a trained model performs, especially on data it has not seen during training.

The choice of evaluation metric depends on the type of machine learning problem and what we consider important in the predictions.

# Regression Evaluation Metric — MAE

## What is MAE?

Mean Absolute Error (MAE) measures the average absolute difference between the actual values and the predicted values of a model.

It tells us, on average, how far the model's predictions are from the actual values.

The equation is:

$$
MAE =
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $n$ = Number of observations

---

## Example

Suppose a trained model makes the following predictions:

| Actual | Predicted |
|---:|---:|
| 10 | 8 |
| 20 | 23 |
| 30 | 27 |

The absolute errors are:

$$
|10-8|=2
$$

$$
|20-23|=3
$$

$$
|30-27|=3
$$

Therefore:

$$
MAE=\frac{2+3+3}{3}
$$

$$
\boxed{MAE=2.67}
$$

This means that the model's predictions are, on average, 2.67 units away from the actual values.

---

## Why Is MAE Useful for Evaluation?

MAE is easy to interpret because it is expressed in the same units as the target variable.

For example, if we are predicting a property measured in a particular unit and:

$$
MAE=0.35
$$

we can interpret this as:

> The model's predictions differ from the actual values by approximately 0.35 units on average.

---

## Important Point

A lower MAE generally indicates better predictive performance.

For example:

$$
MAE=0.25
$$

is generally better than:

$$
MAE=0.80
$$

provided that both values are calculated on the same dataset and under comparable conditions.

---

## MAE as an Evaluation Metric

During evaluation, we do not use MAE to update the model's weights.

Instead, we use it to measure the performance of the already-trained model.

The process is:

$$
\text{Unseen Data}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Predictions}
\rightarrow
\text{MAE}
$$

The resulting MAE helps us understand how accurately the model predicts new data.

---

## Key Takeaway

> MAE tells us the average magnitude of the prediction error in the original units of the target.

A lower MAE generally means that the model's predictions are closer to the actual values.

# Regression Evaluation Metric — MSE

## What is MSE?

Mean Squared Error (MSE) measures the average of the squared differences between the actual values and the predicted values.

It gives more importance to larger prediction errors because the errors are squared.

The equation is:

$$
MSE =
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $n$ = Number of observations

---

## Example

Suppose a trained model makes the following predictions:

| Actual | Predicted | Error | Squared Error |
|---:|---:|---:|---:|
| 10 | 8 | 2 | 4 |
| 20 | 23 | -3 | 9 |
| 30 | 27 | 3 | 9 |

Therefore:

$$
MSE=
\frac{4+9+9}{3}
$$

$$
\boxed{MSE=7.33}
$$

---

## Why Does MSE Give More Importance to Large Errors?

Consider two errors:

$$
Error=2
$$

and

$$
Error=10
$$

After squaring:

$$
2^2=4
$$

$$
10^2=100
$$

The larger error becomes much more influential.

Therefore:

> MSE is more sensitive to large errors and outliers than MAE.

---

## MSE as an Evaluation Metric

After training the model, we can calculate MSE on validation or test data to measure its prediction performance.

The process is:

$$
\text{Unseen Data}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Predictions}
\rightarrow
\text{MSE}
$$

A lower MSE generally indicates better predictive performance when comparing models on the same dataset.

---

## Important Limitation

MSE is expressed in squared units of the target.

For example, if the target is measured in units $U$:

$$
\text{MSE unit}=U^2
$$

Therefore, MSE is useful mathematically and for comparing models, but it can be less intuitive to interpret directly.

This is one reason we also use RMSE.

---

## MAE vs MSE for Evaluation

| Feature | MAE | MSE |
|---|---|---|
| Error treatment | Absolute | Squared |
| Large errors | Less influence | Strong influence |
| Outlier sensitivity | Lower | Higher |
| Units | Same as target | Squared units |
| Interpretation | Easy | Less intuitive |

---

## Key Takeaway

> MSE evaluates the average squared prediction error and gives greater importance to large errors.

A lower MSE generally indicates better predictive performance, but the metric should always be interpreted in the context of the problem.

# Regression Evaluation Metric — RMSE

## What is RMSE?

Root Mean Squared Error (RMSE) is the square root of the Mean Squared Error (MSE).

It measures the typical size of prediction errors while giving more importance to larger errors.

The equation is:

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $n$ = Number of observations

---

## Example

Suppose the model has the following errors:

$$
2,\;-3,\;3
$$

First, square the errors:

$$
2^2=4
$$

$$
(-3)^2=9
$$

$$
3^2=9
$$

Calculate MSE:

$$
MSE=\frac{4+9+9}{3}
$$

$$
MSE=7.33
$$

Now take the square root:

$$
RMSE=\sqrt{7.33}
$$

$$
\boxed{RMSE\approx2.71}
$$

---

## Why Is RMSE Useful for Evaluation?

RMSE has two important properties:

### 1. It penalizes large errors more strongly

Because the errors are squared before averaging:

$$
2^2=4
$$

while:

$$
10^2=100
$$

Therefore, large prediction errors have a much greater influence on RMSE.

### 2. It is expressed in the original units

MSE is expressed in squared units.

Taking the square root brings the result back to the same units as the target.

Therefore:

$$
\boxed{RMSE\text{ is in the same units as the target}}
$$

---

## Interpreting RMSE

Suppose a model has:

$$
RMSE=0.40
$$

This means that the model's prediction errors have a typical magnitude of approximately 0.40 target units, with larger errors having greater influence on the metric.

A lower RMSE generally indicates better predictive performance when comparing models on the same dataset.

---

## MAE vs RMSE

Both MAE and RMSE are expressed in the original units of the target, but they treat errors differently.

| Feature | MAE | RMSE |
|---|---|---|
| Error treatment | Absolute | Squared, then square root |
| Large errors | Less influence | More influence |
| Outlier sensitivity | Lower | Higher |
| Units | Same as target | Same as target |
| Interpretation | Average absolute error | Error measure emphasizing large errors |

---

## What Does MAE vs RMSE Tell Us?

Comparing MAE and RMSE can provide additional information about the model's errors.

If:

$$
MAE \approx RMSE
$$

the errors are relatively consistent and there may not be very large errors.

If:

$$
RMSE \gg MAE
$$

some larger errors are likely having a stronger influence on the RMSE.

Therefore, looking at both metrics can provide more information than looking at only one.

---

## RMSE as an Evaluation Metric

The evaluation process is:

$$
\text{Unseen Data}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Predictions}
\rightarrow
\text{RMSE}
$$

RMSE helps us understand how well the trained model predicts unseen data.

---

## Key Takeaway

> RMSE measures prediction error in the original units of the target while giving greater influence to large errors.

A lower RMSE generally indicates better predictive performance.

# Regression Evaluation Metric — R² (R-squared: coefficient of determination)

## What is R²?

R² (R-squared), also called the coefficient of determination, is a regression evaluation metric that tells us how much of the variation in the target variable is explained by the model.

Unlike MAE, MSE, and RMSE, R² does not directly represent the average size of prediction errors.

Instead, it measures how well the model explains the variation in the target compared with a simple baseline model.

---

## Mathematical Equation

$$
R^2 =
1-
\frac{
\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
}{
\sum_{i=1}^{n}(y_i-\bar{y})^2
}
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $\bar{y}$ = Mean of the actual target values
- $n$ = Number of observations

The numerator represents the model's squared prediction error.

The denominator represents the total variation of the actual target values around their mean.

---

## The Basic Idea

R² compares two things:

### Model

> How much error does our model make?

### Baseline

> How much error would we make if we simply predicted the mean of the target for every observation?

The model is considered useful when it performs better than this simple mean-based baseline.

---

## Understanding R² Values

R² is commonly interpreted as follows:

$$
R^2=1
$$

means the model explains the data perfectly.

For example:

$$
R^2=0.80
$$

can be interpreted as the model explaining approximately 80% of the variation in the target, under the assumptions of this interpretation.

Similarly:

$$
R^2=0.50
$$

means the model explains approximately 50% of the variation in the target.

---

## What Does R² = 0 Mean?

If:

$$
R^2=0
$$

the model performs approximately as well as the baseline that always predicts the mean of the target.

Therefore, the model is not explaining additional variation compared with that baseline.

---

## Can R² Be Negative?

Yes.

Although R² is often between 0 and 1 for a good regression model, it can be negative, especially when evaluated on unseen data.

A negative R² means:

> The model performs worse than simply predicting the mean of the target for every observation.

For example:

$$
R^2=-0.20
$$

means the model is performing worse than the mean-prediction baseline.

---

## R² vs MAE / RMSE

These metrics tell us different things.

| Metric | What does it tell us? |
|---|---|
| MAE | Average absolute prediction error |
| MSE | Average squared prediction error |
| RMSE | Error size with greater emphasis on large errors |
| R² | How much variation in the target is explained by the model |

Therefore, it is useful to report more than one metric.

For example:

$$
MAE=0.36
$$

$$
RMSE=0.55
$$

$$
R^2=0.61
$$

Together, these metrics give us a better understanding of model performance.

---

## R² During Model Evaluation

After training:

$$
\text{Unseen Data}
\rightarrow
\text{Trained Model}
\rightarrow
\text{Predictions}
\rightarrow
R^2
$$

The R² score helps us understand how well the model explains the variation in unseen data.

---

## Key Takeaway

> R² measures how much of the variation in the target variable is explained by the regression model compared with a mean-prediction baseline.

A higher R² generally indicates better explanatory performance, but R² should be considered together with error metrics such as MAE and RMSE.

# Adjusted R²

## What is Adjusted R²?

Adjusted R² is a modified version of R² that takes into account the number of predictors (features) used by a regression model.

R² can increase when additional predictors are added to a model, even when those predictors provide little useful information.

Adjusted R² introduces a penalty for adding unnecessary predictors.

---

## Why Do We Need Adjusted R²?

Suppose we have a regression model:

### Model 1

Uses 5 features:

$$
R^2=0.80
$$

### Model 2

Uses 50 features:

$$
R^2=0.82
$$

R² increased from 0.80 to 0.82.

However, we don't know whether the additional 45 features actually improved the model meaningfully.

Adjusted R² considers the number of predictors and therefore gives us a better way to compare models with different numbers of features.

---

## Mathematical Equation

$$
Adjusted\ R^2
=
1-
(1-R^2)
\frac{n-1}{n-p-1}
$$

Where:

- $R^2$ = R-squared
- $n$ = Number of observations
- $p$ = Number of predictors/features

---

## How Does the Adjustment Work?

When useful predictors are added:

$$
R^2 \uparrow
$$

and Adjusted R² may also increase.

However, if unnecessary predictors are added:

$$
R^2 \uparrow
$$

but Adjusted R² may:

$$
\downarrow
$$

or remain almost unchanged.

Therefore, Adjusted R² discourages adding features that do not provide enough improvement.

---

## R² vs Adjusted R²

| Feature | R² | Adjusted R² |
|---|---|---|
| Measures explained variation | Yes | Yes |
| Increases when features are added | Generally yes | Not necessarily |
| Accounts for number of predictors | No | Yes |
| Useful for comparing models with different feature counts | Limited | More useful |

---

## Important Limitation

Adjusted R² is especially useful in traditional regression when comparing models with different numbers of predictors.

For high-dimensional machine-learning models, such as models using thousands of molecular fingerprints, metrics such as:

- MAE
- RMSE
- R²

on an independent test set are generally more useful for evaluating predictive performance.

---

## Key Takeaway

> Adjusted R² modifies R² by accounting for the number of predictors in the model. It can help determine whether adding additional features actually improves a regression model enough to justify their inclusion.

# Classification Evaluation 

# 01_Confusion Matrix

## What is a Confusion Matrix?

A confusion matrix is a table used to evaluate the performance of a classification model.

It compares:

- The actual class
- The predicted class

For binary classification, we usually have two classes:

$$
0 = Negative
$$

$$
1 = Positive
$$

The confusion matrix tells us not only whether the model was correct, but also what type of mistake it made.

---

## The Four Possible Outcomes

For binary classification, every prediction falls into one of four categories.

### 1. True Positive (TP)

The actual class is positive, and the model predicts positive.

$$
Actual=1,\quad Prediction=1
$$

Example:

> A toxic molecule is correctly predicted as toxic.

$$
\boxed{TP}
$$

---

### 2. True Negative (TN)

The actual class is negative, and the model predicts negative.

$$
Actual=0,\quad Prediction=0
$$

Example:

> A non-toxic molecule is correctly predicted as non-toxic.

$$
\boxed{TN}
$$

---

### 3. False Positive (FP)

The actual class is negative, but the model predicts positive.

$$
Actual=0,\quad Prediction=1
$$

Example:

> A non-toxic molecule is incorrectly predicted as toxic.

This is also called a:

False Alarm

$$
\boxed{FP}
$$

---

### 4. False Negative (FN)

The actual class is positive, but the model predicts negative.

$$
Actual=1,\quad Prediction=0
$$

Example:

> A toxic molecule is incorrectly predicted as non-toxic.

This is also called a:

Miss

$$
\boxed{FN}
$$

---

# Confusion Matrix Structure

The confusion matrix can be represented as:

| | Predicted Negative | Predicted Positive |
|---|---:|---:|
| Actual Negative | TN | FP |
| Actual Positive | FN | TP |

Another way to visualize it:

$$
\begin{array}{c|cc}
& Prediction\ 0 & Prediction\ 1\\
\hline
Actual\ 0 & TN & FP\\
Actual\ 1 & FN & TP
\end{array}
$$

---

## Example

Suppose we evaluate a toxicity classification model on 100 molecules.

The model produces:

- TN = 60
- FP = 10
- FN = 5
- TP = 25

The confusion matrix is:

| | Predicted Non-Toxic | Predicted Toxic |
|---|---:|---:|
| Actual Non-Toxic | 60 | 10 |
| Actual Toxic | 5 | 25 |

Total predictions:

$$
60+10+5+25=100
$$

---

## Why Is the Confusion Matrix Important?

A simple accuracy value might tell us:

> "The model is 85% accurate."

But that does not tell us what kinds of mistakes the model is making.

The confusion matrix shows us:

- How many positives were correctly detected
- How many negatives were correctly detected
- How many false alarms occurred
- How many positive cases were missed

This information is used to calculate other classification metrics.

---

## From Confusion Matrix to Other Metrics

The four values:

$$
TP,\quad TN,\quad FP,\quad FN
$$

are the foundation for several important evaluation metrics:

$$
\boxed{Accuracy}
$$

$$
\boxed{Precision}
$$

$$
\boxed{Recall}
$$

$$
\boxed{F1\text{-}score}
$$

Therefore:

$$
\text{Confusion Matrix}
\rightarrow
\text{Classification Metrics}
$$

---

## Important Terminology

### False Positive

The model says Positive, but the actual class is Negative.

$$
FP:
Actual=0,\ Prediction=1
$$

### False Negative

The model says Negative, but the actual class is Positive.

$$
FN:
Actual=1,\ Prediction=0
$$

These two errors are particularly important because different applications may consider one type of error more serious than the other.

---

## Key Takeaway

> A confusion matrix breaks classification predictions into four categories: True Positive, True Negative, False Positive, and False Negative.

These four values form the foundation for understanding most classification evaluation metrics.

## Python Implementation

We can calculate a confusion matrix using scikit-learn:

from sklearn.metrics import confusion_matrix

y_true = [1, 0, 1, 1, 0, 0]
y_pred = [1, 0, 0, 1, 1, 0]

cm = confusion_matrix(y_true, y_pred)

print(cm)
The output has the structure:

[[TN, FP],
 [FN, TP]]
We can also extract the four values:

TN, FP, FN, TP = cm.ravel()

print("True Negative :", TN)
print("False Positive:", FP)
print("False Negative:", FN)
print("True Positive :", TP)
These four values are the foundation for calculating:

- Accuracy
- Precision
- Recall
- F1-score

# Classification Evaluation — Accuracy

## What is Accuracy?

Accuracy measures the proportion of predictions that the classification model gets correct out of all predictions.

It considers both:

- True Positives (TP)
- True Negatives (TN)

The equation is:

$$
Accuracy =
\frac{TP+TN}
{TP+TN+FP+FN}
$$

---

## Example

Suppose our confusion matrix contains:

- $TP=25$
- $TN=60$
- $FP=10$
- $FN=5$

Total predictions:

$$
25+60+10+5=100
$$

Correct predictions:

$$
TP+TN=25+60=85
$$

Therefore:

$$
Accuracy=\frac{85}{100}
$$

$$
\boxed{Accuracy=0.85}
$$

or:

$$
\boxed{Accuracy=85\%}
$$

This means the model correctly classified 85 out of 100 samples.

---

## Python Implementation

We can calculate accuracy using scikit-learn:

from sklearn.metrics import accuracy_score

y_true = [1, 0, 1, 1, 0, 0]
y_pred = [1, 0, 0, 1, 1, 0]

accuracy = accuracy_score(y_true, y_pred)

print("Accuracy:", accuracy)
To display it as a percentage:

print(f"Accuracy: {accuracy * 100:.2f}%")
---

## When Is Accuracy Useful?

Accuracy can be useful when:

- The classes are reasonably balanced.
- False positives and false negatives have similar importance.
- We want a simple overall measure of correct predictions.

For example:

$$
50\ positive + 50\ negative
$$

is a relatively balanced dataset.

In such a situation, accuracy can provide a useful overall summary.

---

## Important Limitation of Accuracy

Accuracy can be misleading when the classes are highly imbalanced.

Suppose we have 1000 molecules:

- 990 → Non-toxic
- 10 → Toxic

Imagine a model predicts:

> Every molecule is non-toxic.

The model gets:

$$
990
$$

predictions correct.

Therefore:

$$
Accuracy=\frac{990}{1000}=99\%
$$

It looks excellent!

But the model detected:

$$
0/10
$$

toxic molecules.

So despite having 99% accuracy, the model is completely failing to identify the minority class.

This is why we should not automatically consider a high accuracy score to mean that a classification model is good.

---

## Accuracy and Confusion Matrix

Accuracy is calculated directly from the four confusion-matrix values:

$$
TP,\ TN,\ FP,\ FN
$$

Therefore:

$$
\text{Confusion Matrix}
\rightarrow
TP,TN,FP,FN
\rightarrow
\text{Accuracy}
$$

---

## Key Takeaway

> Accuracy is the proportion of all predictions that are correct.

$$
\boxed{
Accuracy=
\frac{TP+TN}
{TP+TN+FP+FN}
}
$$

> Accuracy is useful for balanced classification problems, but it can be misleading when the classes are highly imbalanced.

```python
# Python Function 

def accuracy(actual, predicted):
    correct = 0

    for a, p in zip(actual, predicted):
        if a == p:
            correct += 1

    return correct / len(actual)
```

```python
# Example


actual = [1, 0, 1, 1, 0]
predicted = [1, 0, 0, 1, 1]

result = accuracy(actual, predicted)

print("Accuracy:", result)
```

# Classification Evaluation — Precision

## What is Precision?

Precision measures how reliable the model's positive predictions are.

It answers the question:

> Out of all samples predicted as Positive, how many were actually Positive?

The equation is:

$$
Precision =
\frac{TP}{TP+FP}
$$

Where:

- $TP$ = True Positive
- $FP$ = False Positive

---

## Example

Suppose:

$$
TP=80
$$

$$
FP=20
$$

Then:

$$
Precision=
\frac{80}{80+20}
$$

$$
Precision=0.80
$$

Therefore:

$$
\boxed{Precision=80\%}
$$

This means:

> 80% of the samples predicted as Positive were actually Positive.

---

## Why Does Precision Focus on False Positives?

Precision is concerned with the correctness of Positive predictions.

Therefore, it uses:

$$
TP+FP
$$

because these are all the samples that the model predicted as Positive.

                  Model Prediction
                       Positive
                          ↓
                  ┌───────────────┐
                  │               │
                TP              FP
             Correct          Incorrect
             Positive         Positive
                  │               │
                  └───────┬───────┘
                          ↓
                    TP + FP
Therefore:

$$
Precision=\frac{Correct\ Positive\ Predictions}
{All\ Positive\ Predictions}
$$

---

## High Precision

High precision means that when the model predicts Positive, it is usually correct.

For example:

$$
Precision=0.95
$$

means:

> Approximately 95% of the model's positive predictions were actually positive.

---

## Low Precision

Low precision means the model produces many false positive predictions.

For example:

$$
Precision=0.50
$$

means:

> Only 50% of the model's positive predictions were actually positive.

---

## Key Takeaway

> Precision tells us how trustworthy the model's Positive predictions are.

$$
\boxed{
Precision=\frac{TP}{TP+FP}
}
$$

Precision focuses on False Positives.

```python
# Python Function

def precision(actual, predicted):
    tp = 0
    fp = 0

    for a, p in zip(actual, predicted):
        if p == 1:
            if a == 1:
                tp += 1
            else:
                fp += 1

    return tp / (tp + fp)

# Example

actual = [1, 0, 1, 1, 0, 0]
predicted = [1, 1, 1, 0, 0, 1]

result = precision(actual, predicted)

print("Precision:", result)
```

# Classification Evaluation — Recall

## What is Recall?

Recall measures how well the model identifies the actual Positive cases.

It answers the question:

> Out of all the samples that are actually Positive, how many did the model correctly identify?

The equation is:

$$
Recall =
\frac{TP}{TP+FN}
$$

Where:

- $TP$ = True Positive
- $FN$ = False Negative

---

## Example

Suppose there are 100 actually Positive samples.

The model correctly identifies 80 of them:

$$
TP=80
$$

The remaining 20 are incorrectly predicted as Negative:

$$
FN=20
$$

Therefore:

$$
Recall=
\frac{80}{80+20}
$$

$$
Recall=0.80
$$

Therefore:

$$
\boxed{Recall=80\%}
$$

This means the model successfully identified 80% of all the actual Positive cases.

---

## Why Does Recall Focus on False Negatives?

Recall is concerned with finding the actual Positive cases.

Therefore, it uses:

$$
TP+FN
$$

because these represent all the samples that are actually Positive.

                 Actual Positive
                       ↓
              ┌─────────────────┐
              │                 │
             TP                FN
          Found correctly    Missed
              │                 │
              └────────┬────────┘
                       ↓
                    TP + FN
Therefore:

$$
Recall=
\frac{Correctly\ Identified\ Positives}
{All\ Actual\ Positives}
$$

---

## High Recall

High recall means the model finds most of the actual Positive cases.

For example:

$$
Recall=0.95
$$

means:

> The model identified approximately 95% of all actual Positive cases.

---

## Low Recall

Low recall means the model is missing many actual Positive cases.

For example:

$$
Recall=0.50
$$

means:

> The model identified only 50% of the actual Positive cases.

---

## Precision vs Recall

These metrics ask different questions.

### Precision

> When the model says Positive, how often is it correct?

$$
Precision=\frac{TP}{TP+FP}
$$

Focus:

$$
\boxed{FP}
$$

---

### Recall

> Of all the actual Positive cases, how many did the model find?

$$
Recall=\frac{TP}{TP+FN}
$$

Focus:

$$
\boxed{FN}
$$

---

## Simple Memory Trick

### Precision

> Predicted Positive → How many were correct?

$$
TP/(TP+FP)
$$

### Recall

> Actual Positive → How many did we find?

$$
TP/(TP+FN)
$$

---

## Key Takeaway

> Recall tells us how effectively the model detects the actual Positive cases.

$$
\boxed{
Recall=\frac{TP}{TP+FN}
}
$$

Recall focuses on False Negatives.

```python
# Python Function

def recall(actual, predicted):
    tp = 0
    fn = 0

    for a, p in zip(actual, predicted):
        if a == 1:
            if p == 1:
                tp += 1
            else:
                fn += 1

    return tp / (tp + fn)


# Example

actual = [1, 0, 1, 1, 0, 1]
predicted = [1, 1, 1, 0, 0, 1]

result = recall(actual, predicted)

print("Recall:", result)
```

# Classification Evaluation — F1-Score

## What is F1-Score?

F1-score combines Precision and Recall into a single metric.

It is particularly useful when we want a balance between:

- Correct Positive predictions → Precision
- Finding actual Positive cases → Recall

The equation is:

$$
F1 =
2\times
\frac{Precision\times Recall}
{Precision+Recall}
$$

---

## Why Do We Need F1-Score?

Sometimes a model can have:

- High Precision but low Recall
- High Recall but low Precision

Looking at only one of them may not give us the complete picture.

F1-score combines both.

---

## Example

Suppose:

$$
Precision=0.80
$$

and:

$$
Recall=0.60
$$

Then:

$$
F1=
2\times
\frac{0.80\times0.60}
{0.80+0.60}
$$

$$
F1=
2\times\frac{0.48}{1.40}
$$

$$
\boxed{F1\approx0.686}
$$

Therefore:

$$
\boxed{F1\approx68.6\%}
$$

---

## Why Does F1 Use Harmonic Mean?

F1-score uses the harmonic mean rather than the ordinary arithmetic mean.

The harmonic mean gives more importance to the lower value.

For example:

$$
Precision=0.90
$$

$$
Recall=0.10
$$

The ordinary average would be:

$$
\frac{0.90+0.10}{2}=0.50
$$

But F1-score is:

$$
F1=
2\times
\frac{0.90\times0.10}
{0.90+0.10}
$$

$$
F1=0.18
$$

So F1 does not allow a very high Precision to hide a very low Recall.

---

## Relationship Between the Metrics

The three metrics focus on different aspects:

### Precision

> Out of the Positive predictions, how many were actually Positive?

$$
Precision=\frac{TP}{TP+FP}
$$

### Recall

> Out of the actual Positive cases, how many did the model find?

$$
Recall=\frac{TP}{TP+FN}
$$

### F1-score

> How well are Precision and Recall balanced?

$$
F1=
2\times
\frac{Precision\times Recall}
{Precision+Recall}
$$

---

## Range of F1-Score

F1-score ranges from:

$$
0\leq F1\leq1
$$

A value close to:

$$
1
$$

indicates a strong balance between Precision and Recall.

A value close to:

$$
0
$$

indicates poor performance.

---

## Python Function

We can calculate F1-score manually:

def f1_score_manual(precision, recall):

    return 2 * (precision * recall) / (precision + recall)
    
Example:

precision = 0.80

recall = 0.60

f1 = f1_score_manual(precision, recall)

print("F1-score:", f1)



## Scikit-Learn Function

In practical machine-learning projects, we can use:

from sklearn.metrics import f1_score

f1 = f1_score(actual, predicted)

print("F1-score:", f1)


## Key Takeaway

> F1-score provides a single measure that balances Precision and Recall.

$$
\boxed{
F1=
2\times
\frac{Precision\times Recall}
{Precision+Recall}
}
$$

A high F1-score generally requires both Precision and Recall to be reasonably high.

# Classification Evaluation — Specificity

## What is Specificity?

Specificity measures how well the model identifies the actual Negative cases.

It answers the question:

> Out of all the samples that are actually Negative, how many did the model correctly identify as Negative?

The equation is:

$$
Specificity =
\frac{TN}{TN+FP}
$$

Where:

- $TN$ = True Negative
- $FP$ = False Positive

---

## Example

Suppose we have:

$$
TN=90
$$

$$
FP=10
$$

Then:

$$
Specificity=
\frac{90}{90+10}
$$

$$
Specificity=0.90
$$

Therefore:

$$
\boxed{Specificity=90\%}
$$

This means:

> The model correctly identified 90% of all actual Negative cases.

---

## Why Does Specificity Use TN and FP?

Specificity focuses only on the samples that are actually Negative.

Those samples can have only two outcomes:

Therefore:

$$
Actual\ Negative = TN+FP
$$

And:

$$
Specificity=
\frac{TN}{TN+FP}
$$

---

## Specificity vs Recall

These two metrics look at opposite actual classes.

### Recall

Looks at Actual Positive cases:

$$
Recall=\frac{TP}{TP+FN}
$$

### Specificity

Looks at Actual Negative cases:

$$
Specificity=\frac{TN}{TN+FP}
$$

So:

# Recall

   ↓

Actual Positive

   ↓

How many did we correctly find?



# Specificity

   ↓

Actual Negative

   ↓

How many did we correctly identify?



## Relationship with False Positive Rate

Specificity and False Positive Rate are complementary:

$$
FPR=\frac{FP}{TN+FP}
$$

Therefore:

$$
\boxed{Specificity=1-FPR}
$$

---

## Python Function

```python

def specificity(actual, predicted):
    tn = 0
    fp = 0

    for a, p in zip(actual, predicted):
        if a == 0:
            if p == 0:
                tn += 1
            else:
                fp += 1

    return tn / (tn + fp)
```

```python
# Example


actual = [1, 0, 1, 0, 0, 1]
predicted = [1, 0, 0, 0, 1, 1]

result = specificity(actual, predicted)

print("Specificity:", result)
```

## Scikit-Learn Function

Scikit-learn does not require a separate specificity() function.

We can calculate it using the confusion matrix:

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(actual, predicted)

TN, FP, FN, TP = cm.ravel()

specificity = TN / (TN + FP)

print("Specificity:", specificity)
```

## Key Takeaway

> Specificity tells us how well the model identifies the actual Negative cases.

$$
\boxed{
Specificity=\frac{TN}{TN+FP}
}
$$

Specificity focuses on False Positives.

# Classification Evaluation — ROC Curve and AUC

## What is ROC?

ROC stands for Receiver Operating Characteristic.

The ROC curve shows how the classification model behaves when we change the probability threshold used to convert probabilities into class predictions.

A binary classification model may produce:

$$
Probability = 0.87
$$

If we use:

$$
Threshold=0.5
$$

then:

$$
0.87 \geq 0.5
\Rightarrow Prediction=1
$$

But if we change the threshold, the predicted class can change.

Therefore, ROC analysis evaluates the model across many different thresholds rather than using only one threshold such as 0.5.

---

# True Positive Rate (TPR)

The True Positive Rate is the same as Recall.

$$
TPR =
\frac{TP}{TP+FN}
$$

It tells us:

> Out of all actual Positive samples, how many did the model correctly identify?

---

# False Positive Rate (FPR)

The False Positive Rate measures how many actual Negative samples were incorrectly predicted as Positive.

$$
FPR =
\frac{FP}{FP+TN}
$$

It tells us:

> Out of all actual Negative samples, how many were incorrectly classified as Positive?

---

# ROC Curve

The ROC curve plots:

$$
TPR
$$

against:

$$
FPR
$$

at different classification thresholds.

A model whose ROC curve stays closer to the top-left corner generally has better ability to distinguish between the two classes.

---

# What Happens When We Change the Threshold?

Suppose the model produces these probabilities:

| Molecule | Probability |
|---|---:|
| A | 0.95 |
| B | 0.80 |
| C | 0.60 |
| D | 0.40 |
| E | 0.20 |

### Threshold = 0.5

Predictions:

$$
[1,1,1,0,0]
$$

### Threshold = 0.8

Predictions:

$$
[1,1,0,0,0]
$$

Changing the threshold changes the number of:

- TP
- FP
- TN
- FN

Therefore, it also changes:

- TPR
- FPR
- Precision
- Recall

The ROC curve examines this behavior across many thresholds.

---

# What is AUC?

AUC stands for Area Under the ROC Curve.

It summarizes the ROC curve into a single value.

$$
\boxed{AUC = Area\ Under\ the\ ROC\ Curve}
$$

AUC generally ranges from:

$$
0\leq AUC\leq1
$$

---

## Interpreting AUC

### AUC = 1.0

The model has perfect discrimination between the two classes.

### AUC ≈ 0.5

The model performs approximately like random ranking.

### AUC < 0.5

The model is ranking the classes worse than random.

In practice, a value below 0.5 can sometimes indicate that the predictions or class interpretation may be reversed.

---

# What Does AUC Actually Tell Us?

AUC primarily measures the model's ability to rank Positive samples above Negative samples across thresholds.

A higher AUC generally means better discrimination between the two classes.

Importantly:

> AUC does not tell us which threshold should be used for the final classification.

The choice of threshold depends on the application and the relative importance of different types of errors.

---

# Python Function

We can calculate ROC-AUC using scikit-learn:

```python
from sklearn.metrics import roc_auc_score

actual = [1, 0, 1, 1, 0, 0]

probabilities = [0.90, 0.20, 0.80, 0.70, 0.30, 0.10]

auc = roc_auc_score(actual, probabilities)

print("ROC-AUC:", auc)
```

Important: AUC should be calculated using the model's probabilities/scores, not only the final 0/1 predictions.

---

# Key Takeaway

> ROC shows how the model's True Positive Rate and False Positive Rate change across different probability thresholds.

> AUC summarizes the ROC curve and measures the model's ability to distinguish between Positive and Negative samples.

$$
\boxed{
TPR=\frac{TP}{TP+FN}
}
$$

$$
\boxed{
FPR=\frac{FP}{FP+TN}
}
$$

$$
\boxed{
AUC=Area\ Under\ ROC\ Curve
}
$$

# Classification Evaluation — Precision-Recall Curve

## What is a Precision-Recall Curve?

A Precision-Recall (PR) Curve shows how Precision and Recall change when we change the classification threshold.

A binary classification model produces a probability:

$$
0 \leq P \leq 1
$$

We choose a threshold to convert this probability into a class prediction.

For example:

$$
P \geq 0.5 \Rightarrow Class=1
$$

Changing the threshold changes the number of:

- TP
- FP
- FN
- TN

Therefore, it also changes:

- Precision
- Recall

The Precision-Recall curve shows this relationship across different thresholds.

---

## What is Plotted?

The PR curve plots:

$$
Precision
$$

against:

$$
Recall
$$

Usually:

- X-axis → Recall
- Y-axis → Precision


## Why Does Precision Change with Recall?

Suppose our model predicts probabilities for several molecules.

If we use a high threshold, the model becomes more selective about calling something Positive.

For example:

$$
Threshold=0.9
$$

Only molecules with very high predicted probability are classified as Positive.

This may result in:

- Fewer False Positives
- Higher Precision
- Lower Recall

If we lower the threshold:

$$
Threshold=0.3
$$

More molecules are classified as Positive.

This may result in:

- More True Positives
- Higher Recall
- Potentially more False Positives
- Lower Precision

Therefore, Precision and Recall often have a trade-off.

---

# Precision-Recall Trade-off

A typical relationship can be represented as:

$$
Recall \uparrow
\Rightarrow
Precision\ may\downarrow
$$

and:

$$
Precision \uparrow
\Rightarrow
Recall\ may\downarrow
$$

The exact relationship depends on the model and dataset.

---

# Why is PR Curve Useful for Imbalanced Data?

Consider a dataset containing:

$$
990\ Negative
$$

and:

$$
10\ Positive
$$

The classes are highly imbalanced.

In such a situation, accuracy can sometimes give a misleadingly high value.

Precision and Recall focus directly on the Positive class.

Therefore, a Precision-Recall curve can provide a more informative view of Positive-class performance when the Positive class is rare.

---

# Average Precision (AP)

The Average Precision (AP) score summarizes the Precision-Recall curve into a single value.

A higher AP generally indicates better performance across the range of recall values.

It is commonly used as a summary metric for Precision-Recall performance.

---

# Python Implementation

We can calculate the Precision-Recall curve using scikit-learn:

```python
from sklearn.metrics import precision_recall_curve

actual = [1, 0, 1, 1, 0, 0]

probabilities = [0.90, 0.20, 0.80, 0.70, 0.30, 0.10]

precision, recall, thresholds = precision_recall_curve(
    actual,
    probabilities
)

print("Precision:", precision)
print("Recall:", recall)
print("Thresholds:", thresholds)
```

## Calculate Average Precision

```python
from sklearn.metrics import average_precision_score

ap = average_precision_score(actual, probabilities)

print("Average Precision:", ap)
```

```python
## Plot the Precision-Recall Curve

import matplotlib.pyplot as plt
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(
    actual,
    probabilities
)

plt.plot(recall, precision)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")

plt.show()
```

# ROC Curve vs Precision-Recall Curve

| ROC Curve | Precision-Recall Curve |
|---|---|
| TPR vs FPR | Precision vs Recall |
| Shows class discrimination across thresholds | Focuses on Positive-class performance |
| Often useful for balanced datasets | Particularly useful for imbalanced datasets |
| Uses TP, FN, FP, TN | Mainly focuses on TP, FP, FN |

---

# Key Takeaway

> A Precision-Recall curve shows the relationship between Precision and Recall as the classification threshold changes.

It is particularly useful when the Positive class is relatively rare.

$$
\boxed{
PR\ Curve = Precision\ vs\ Recall
}
$$

The choice of threshold should depend on the application and on whether False Positives or False Negatives are more important.

# Classification Evaluation — Log Loss

## What is Log Loss?

Log Loss, also called Cross-Entropy Loss or Logarithmic Loss, measures how well the predicted probabilities match the actual class labels.

It evaluates not only whether the model predicted the correct class, but also how confident the model was in its prediction.

For binary classification:

$$
LogLoss =
-\frac{1}{n}
\sum_{i=1}^{n}
\left[
y_i\log(\hat{p}_i)
+
(1-y_i)\log(1-\hat{p}_i)
\right]
$$

Where:

- $y_i$ = Actual class ($0$ or $1$)
- $\hat{p}_i$ = Predicted probability of class 1
- $n$ = Number of observations

---

## Why Does Probability Matter?

Suppose the actual class is:

$$
y=1
$$

Consider two models:

### Model A

$$
P(1)=0.90
$$

### Model B

$$
P(1)=0.51
$$

Both models would predict:

$$
Class=1
$$

if the threshold is 0.5.

Therefore, their accuracy is the same for this sample.

However, Model A is much more confident and assigns a much higher probability to the correct class.

Log Loss can distinguish between these predictions.

---

## What Happens When the Model Is Very Confident and Correct?

Suppose:

$$
y=1
$$

and:

$$
\hat{p}=0.99
$$

Then:

$$
-\log(0.99)
$$

is very small.

Therefore:

$$
\boxed{Low\ LogLoss}
$$

This is a good prediction.

---

## What Happens When the Model Is Very Confident but Wrong?

Suppose:

$$
y=1
$$

but:

$$
\hat{p}=0.01
$$

The model is saying:

> "There is only a 1% probability that this is Positive."

But the actual class is Positive.

The loss becomes:

$$
-\log(0.01)
$$

which is very large.

Therefore:

$$
\boxed{High\ LogLoss}
$$

This strongly penalizes confident incorrect predictions.

---

## Why Is Log Loss Useful as an Evaluation Metric?

Accuracy only considers the final class prediction.

For example:

$$
0.51 \rightarrow Class\ 1
$$

and:

$$
0.99 \rightarrow Class\ 1
$$

Both are simply counted as correct.

Log Loss uses the probability itself, so it can distinguish between them.

Therefore:

> Accuracy evaluates the final class prediction, while Log Loss evaluates the quality of the predicted probabilities.

---

## Lower Is Better

For Log Loss:

$$
\boxed{Lower\ LogLoss = Better}
$$

A model with:

$$
LogLoss=0.20
$$

generally has better probabilistic predictions than a model with:

$$
LogLoss=0.80
$$

when evaluated on the same data.

---

## Python Function

We can calculate Log Loss using scikit-learn:

```python
from sklearn.metrics import log_loss

actual = [1, 0, 1, 1, 0]

probabilities = [0.90, 0.20, 0.80, 0.70, 0.10]

loss = log_loss(actual, probabilities)

print("Log Loss:", loss)
```

Therefore, changing the probability from:

$$
0.51 \rightarrow 0.99
$$

can change Log Loss significantly even though both may produce the same class prediction.

---

## Key Takeaway

> Log Loss evaluates the quality and confidence of predicted probabilities, strongly penalizing confident incorrect predictions.

$$
\boxed{Lower\ LogLoss = Better}
$$

# Which Evaluation Metric Should We Use for Which Problem?

## Regression Metrics

| Metric | When to use |
|---|---|
| MAE | When you want the average prediction error in the original units and want to treat errors more equally. |
| MSE | When you want to penalize large errors more strongly than small errors. |
| RMSE | When you want the error in the original units while still giving more importance to large errors. |
| R² | When you want to know how well the model explains the variation in the target variable. |
| Adjusted R² | When comparing regression models with different numbers of predictors/features. |

---

## Classification Metrics

| Metric | When to use |
|---|---|
| Confusion Matrix | When you want to see the detailed breakdown of TP, TN, FP, and FN. |
| Accuracy | When the classes are reasonably balanced and you want overall correctness. |
| Precision | When False Positives are costly and positive predictions need to be reliable. |
| Recall | When False Negatives are costly and you don't want to miss actual positives. |
| F1-score | When you need a balance between Precision and Recall, especially with imbalanced classes. |
| Specificity | When correctly identifying the Negative class is particularly important. |
| ROC-AUC | When you want to measure overall class discrimination across different thresholds. |
| PR-AUC / Average Precision | When the Positive class is rare or the dataset is highly imbalanced. |
| Log Loss | When you want to evaluate the quality and confidence of predicted probabilities. |

---

# 🧠 Quick Memory

## Regression

MAE       → Average error

MSE       → Penalize large errors

RMSE      → Error in original units + large-error sensitivity

R²        → Explained variation

Adjusted R² → Explained variation + feature penalty

## Classification

Accuracy     → Overall correctness

Precision    → Avoid False Positives

Recall       → Avoid False Negatives

F1-score     → Balance Precision + Recall

Specificity  → Identify Negatives

ROC-AUC      → Overall class discrimination

PR-AUC       → Positive class / imbalanced data

Log Loss     → Probability quality

> Important: In a real project, we usually use multiple complementary metrics rather than relying on only one metric.
