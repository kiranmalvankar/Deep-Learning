# 7. What is a Kernel / Filter?

A kernel (or filter) is a small matrix of learnable values.

For example, a simple 3 × 3 filter:

$$
K=
\begin{bmatrix}
1 & 0 & -1\\
1 & 0 & -1\\
1 & 0 & -1
\end{bmatrix}
$$

The actual CNN filters are learned from the training data rather than manually specified.

---

# 8. How Does the Filter Work?

Suppose part of an image is:

$$
\begin{bmatrix}
1 & 2 & 3\\
4 & 5 & 6\\
7 & 8 & 9
\end{bmatrix}
$$

and the filter is:

$$
\begin{bmatrix}
1 & 0 & -1\\
1 & 0 & -1\\
1 & 0 & -1
\end{bmatrix}
$$

Multiply corresponding values:

$$
(1)(1)+(2)(0)+(3)(-1)
$$

$$
+(4)(1)+(5)(0)+(6)(-1)
$$

$$
+(7)(1)+(8)(0)+(9)(-1)
$$

Then sum them.

This produces one value in the output feature map.

The filter then moves to another location and repeats the calculation.

# 36. Parameter Sharing

One of the important ideas in CNNs is parameter sharing.

Suppose a 3 × 3 filter is used across an image.

The same filter weights are applied at different locations.


Therefore, the network does not need a completely different set of weights for every image location.

This greatly reduces the number of parameters compared with a fully connected approach.

# 48. CNN vs ANN Summary

| Feature | ANN | CNN |
|---|---|---|
| Main structure | Fully connected layers | Convolution + pooling + dense layers |
| Spatial structure | Not explicitly exploited | Explicitly exploited |
| Local connectivity | No | Yes |
| Parameter sharing | No | Yes |
| Typical use | Tabular/general feature data | Images/spatial data |
| Feature extraction | Dense layers | Convolutional filters |
| Weight sharing | No | Yes |
