
## 1. What is Backpropagation?

Backpropagation is the process used by a neural network to determine how much each weight and bias contributed to the final prediction error.

It then uses this information to calculate gradients, which tell us how the parameters should be changed to reduce the error.

In simple terms:

> Backpropagation tells the neural network how to correct its weights based on the error it made.

---

## 2. Why Do We Need Backpropagation?

During forward propagation, information moves from the input toward the output:

The model then compares the prediction with the actual value and calculates the loss.

For example:

$$
Actual = 1
$$

$$
Prediction = 0.3
$$

Therefore, the model has made an error.

Now we have a problem:

> Which weight should we change, and by how much?

A neural network may contain thousands or millions of parameters.

Backpropagation solves this problem by calculating the gradient of the loss with respect to each parameter.

## 4. Forward Propagation vs Backpropagation

### Forward Propagation

Information moves:

$$
Input \rightarrow Output
$$

Its purpose is to produce a prediction.

### Backpropagation

Information about the error moves backward:

$$
Output \rightarrow Input
$$

Its purpose is to determine how the parameters contributed to the error.

Therefore:


Forward Propagation

→ "What prediction did I make?"

Backpropagation

→ "How should I change my parameters to improve?"

## 6. What Does Backpropagation Calculate?

Backpropagation calculates derivatives such as:

$$
\frac{\partial L}{\partial w}
$$

This means:

> How much does the loss change when a particular weight changes?

Where:

- $L$ = Loss
- $w$ = Weight

This derivative is called a gradient.

For example:

$$
\frac{\partial L}{\partial w}=0.8
$$

means that changing this weight has a measurable effect on the loss.

The sign and magnitude of the gradient provide information about the direction and sensitivity of the change.

---

## 7. Backpropagation Does NOT Directly Update the Weights

This is an important distinction.

Backpropagation calculates the gradients.

An optimization method such as Gradient Descent uses those gradients to update the parameters.

So:

$$
\boxed{
Backpropagation \neq Weight\ Update
}
$$

Instead:

$$
\boxed{
Backpropagation \rightarrow Gradients
}
$$

and:

$$
\boxed{
Optimizer \rightarrow Parameter\ Update
}
$$

# Python Function — Basic Backpropagation Update

Backpropagation calculates the gradient of the loss with respect to a weight:

$$
\frac{\partial L}{\partial w}
$$

The weight can then be updated using:

$$
w_{new}
=
w_{old}
-
\eta
\frac{\partial L}{\partial w}
$$

Where:

- $w$ = Weight
- $\eta$ = Learning rate
- $L$ = Loss
- $\frac{\partial L}{\partial w}$ = Gradient of the loss with respect to the weight


```python
### Python Function

def update_weight(weight, gradient, learning_rate):
    
    new_weight = weight - learning_rate * gradient
    
    return new_weight
```

```python
# Example

weight = 0.5
gradient = 0.8
learning_rate = 0.1

new_weight = update_weight(
    weight,
    gradient,
    learning_rate
)

print("Updated weight:", new_weight)
```
