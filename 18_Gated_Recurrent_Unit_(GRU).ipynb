# 18. Gated Recurrent Unit (GRU)

## 1. What is GRU?

Gated Recurrent Unit (GRU) is a type of recurrent neural network architecture designed to capture dependencies in sequential data while using a simpler structure than LSTM.

GRU was introduced as a more streamlined alternative to LSTM.

In simple words:

> GRU uses gates to decide what information to keep and what information to replace, but it does this with a simpler architecture than LSTM.

# 6. Update Gate

The update gate determines how much of the previous hidden state should be retained versus how much new information should be incorporated.

The equation is:

$$
\boxed{
z_t =
\sigma(W_zx_t+U_zh_{t-1}+b_z)
}
$$

Where:

- $z_t$ = Update gate
- $x_t$ = Current input
- $h_{t-1}$ = Previous hidden state
- $W_z$ = Input-to-gate weights
- $U_z$ = Hidden-to-gate weights
- $b_z$ = Bias
- $\sigma$ = Sigmoid

The sigmoid produces values between:

$$
0 \leq z_t \leq 1
$$

---

# 7. Intuition Behind the Update Gate

Think of the update gate as asking:

> "How much of my previous information should I keep?"

Conceptually:


The exact interpretation depends on the mathematical convention used for the final state equation, so the safest way to remember it is:

> The update gate controls the balance between the previous hidden state and the candidate hidden state.

---

# 8. Reset Gate

The reset gate controls how much of the previous hidden state should influence the candidate state.

The equation is:

$$
\boxed{
r_t =
\sigma(W_rx_t+U_rh_{t-1}+b_r)
}
$$

Where:

- $r_t$ = Reset gate
- $x_t$ = Current input
- $h_{t-1}$ = Previous hidden state
- $W_r$ = Input-to-gate weights
- $U_r$ = Hidden-to-gate weights
- $b_r$ = Bias

---

# 9. Intuition Behind Reset Gate

The reset gate asks:

> "How much of the previous information should I use when creating the new candidate?"

# 10. Candidate Hidden State

GRU creates a candidate hidden state.

A common formulation is:

$$
\boxed{
\tilde{h}_t =
\tanh
\left(
W_hx_t+
U_h(r_t\odot h_{t-1})
+b_h
\right)
}
$$

Where:

- $\tilde{h}_t$ = Candidate hidden state
- $r_t$ = Reset gate
- $h_{t-1}$ = Previous hidden state
- $\odot$ = Element-wise multiplication

The reset gate determines how much previous information participates in creating the candidate.

---

# 11. Final Hidden State

The new hidden state combines the previous hidden state and candidate hidden state.

A common formulation is:

$$
\boxed{
h_t =
(1-z_t)\odot h_{t-1}
+
z_t\odot\tilde{h}_t
}
$$

Here:


Some references use the opposite convention for the update gate. The equations may therefore look different while describing the same underlying mechanism.

---

# 16. Why Does GRU Help With Long-Term Dependencies?

The gating mechanism provides a pathway for information to persist across time steps.

Conceptually:


The update mechanism can allow useful information to persist instead of completely replacing the previous state at every step.

---

# 17. GRU and Vanishing Gradients

Traditional RNNs can have difficulty propagating gradients through long sequences.

GRU's gated state-update mechanism can provide a more effective path for information and gradients.

Conceptually:


GRU does not magically eliminate all gradient problems, but it was designed to improve the handling of long-term dependencies.

# 36. GRU Advantages

1. Simpler than LSTM

2. Fewer parameters than many comparable LSTM
   configurations

3. Can model long-term dependencies

4. Often computationally efficient

5. Useful for sequential data

6. Can be applied to SMILES sequences


---

# 37. GRU Limitations

1. Still sequential and therefore less parallelizable
   than Transformer architectures

2. Can still struggle with extremely long sequences

3. Performance depends on architecture and
   hyperparameters

4. Less expressive memory structure than LSTM
   in some situations

5. Not automatically better than LSTM
