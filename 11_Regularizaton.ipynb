# 📓 10. Regularization

## 1. What is Regularization?

Regularization is a technique used to reduce overfitting by preventing the model from becoming unnecessarily complex.

The main goal is:

$$
\boxed{\text{Reduce Overfitting} \rightarrow \text{Improve Generalization}}
$$

---

## Why Do We Need Regularization?

We already learned that an overly complex model can overfit.

Regularization adds a constraint that discourages the model from relying too heavily on unnecessary complexity.

---

# Simple Analogy

Imagine studying molecular data.

A model could try to explain every tiny variation in the training molecules.

Some variations may represent real chemistry, but some may simply be noise.

Regularization encourages the model to focus more on the important general patterns rather than fitting every small detail.

# 2. L1 Regularization (Lasso)

## What is L1 Regularization?

L1 Regularization adds a penalty based on the absolute values of the model's weights.

The regularized objective becomes:

$$
\boxed{
L_{total}
=
L_{loss}
+
\lambda\sum_i |w_i|
}
$$

Where:

- $L_{loss}$ = Original model loss
- $w_i$ = Model weights
- $\lambda$ = Regularization strength
- $\sum_i |w_i|$ = Sum of the absolute values of the weights

---

## Why Add This Penalty?

Without regularization, the model only tries to minimize prediction error:

$$
L_{total}=L_{loss}
$$

With L1 regularization:

$$
L_{total}
=
L_{loss}
+
\lambda\sum_i |w_i|
$$

Now the model has two objectives:

The model therefore has to find a balance between prediction accuracy and model complexity.

---

### Large λ

$$

$$
\boxed{

}
$$

Because the penalty uses the L1 norm:

# 5. Dropout

Dropout is a regularization technique used in neural networks to reduce overfitting.

During training, Dropout randomly turns off a fraction of neurons in a layer.

# 6. Early Stopping

## What is Early Stopping?

Early Stopping is a regularization technique that stops training when the model stops improving on the validation data.

The purpose is to prevent the model from continuing to learn the training data too specifically.

---

## Why Do We Need Early Stopping?

During training, we may see:

Initially, both losses decrease.

Later, the training loss continues decreasing, but validation loss starts increasing.

This indicates that the model may be overfitting.

Early Stopping allows us to stop around the point where validation performance was best.

---

# Example

Suppose we train for several epochs:

| Epoch | Training Loss | Validation Loss |
|---:|---:|---:|
| 1 | 0.80 | 0.85 |
| 2 | 0.65 | 0.70 |
| 3 | 0.52 | 0.60 |
| 4 | 0.42 | 0.52 |
| 5 | 0.35 | 0.47 |
| 6 | 0.30 | 0.45 |
| 7 | 0.25 | 0.46 |
| 8 | 0.20 | 0.50 |
| 9 | 0.16 | 0.56 |

Epoch 6 has the best validation loss.

After that:

Training Loss ↓

Validation Loss ↑


So continuing to train may increase overfitting.

---

# Patience

We normally don't stop the model immediately after one epoch without improvement.

Instead, we use patience.

For example:

* patience = 3


This means the model can wait for 3 consecutive epochs without improvement before stopping.

Example:

This prevents the model from stopping because of a small temporary fluctuation.

---

# What Does Early Stopping Monitor?

We can monitor a validation metric such as:

- Validation Loss
- Validation MAE
- Validation Accuracy
- Validation R²


The direction depends on the metric.

### For loss / MAE

Lower is better:

### For accuracy
