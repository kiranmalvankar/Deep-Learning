# 1. Gradient — The Direction of Learning

## What is a Gradient?

A gradient tells us how much the loss changes when a model parameter changes.

For a weight $w$, we write:

$$
\frac{\partial L}{\partial w}
$$

This means:

> How much does the loss $L$ change when the weight $w$ changes?

The gradient gives us two important pieces of information:

1. Magnitude → how strongly the weight affects the loss
2. Direction → whether increasing the weight increases or decreases the loss

---

## Simple Example

Suppose the current weight is:

$$
w=2
$$

and the gradient is:

$$
\frac{\partial L}{\partial w}=0.5
$$

The positive gradient tells us that increasing the weight would increase the loss.

Therefore, to reduce the loss, we should move the weight in the opposite direction.

This is why the update equation uses a minus sign:

$$
w_{new}
=
w_{old}
-
\eta
\frac{\partial L}{\partial w}
$$

---

## Example of a Weight Update

Suppose:

$$
w=2
$$

$$
\frac{\partial L}{\partial w}=0.5
$$

and:

$$
\eta=0.1
$$

Then:

$$
w_{new}
=
2-(0.1)(0.5)
$$

$$
w_{new}=2-0.05
$$

$$
\boxed{w_{new}=1.95}
$$

The weight moved from:

$$
2\rightarrow1.95
$$

because the gradient indicated that moving in the negative direction would reduce the loss.

---

# What If the Gradient Is Negative?

Suppose:

$$
\frac{\partial L}{\partial w}=-0.5
$$

Then:

$$
w_{new}
=
2-(0.1)(-0.5)
$$

$$
w_{new}=2+0.05
$$

$$
\boxed{w_{new}=2.05}
$$

So the weight moves upward.

This is why the sign of the gradient is important.

# What Does the Gradient Magnitude Tell Us?

Suppose we have two situations.

### Case 1

$$
\frac{\partial L}{\partial w}=0.1
$$

### Case 2

$$
\frac{\partial L}{\partial w}=2.0
$$

The second gradient is much larger.

This means the loss is much more sensitive to that weight.

Therefore:

The learning rate also controls the final size of the update.

---

# Gradient for Every Weight

A neural network contains many weights.

For example:

$$
w_1,w_2,w_3,\ldots,w_n
$$

Backpropagation calculates a gradient for each one:

$$
\frac{\partial L}{\partial w_1},
\frac{\partial L}{\partial w_2},
\frac{\partial L}{\partial w_3},
\ldots
$$

So we can update every parameter in the network.

# 3. Chain Rule

## What is the Chain Rule?

The Chain Rule is a calculus rule used to calculate how a change in one variable affects another variable through intermediate variables.

In a neural network, information passes through multiple steps:

Because the weight affects the loss through several intermediate steps, we need the Chain Rule to calculate its overall effect.

---

## Simple Example

Suppose:

$$
y = u^2
$$

and:

$$
u = 3x
$$

Here, $x$ affects $y$ through $u$:

We want to calculate:

$$
\frac{dy}{dx}
$$

Using the Chain Rule:

$$
\boxed{
\frac{dy}{dx}
=
\frac{dy}{du}
\times
\frac{du}{dx}
}
$$

Now:

$$
\frac{dy}{du}=2u
$$

and:

$$
\frac{du}{dx}=3
$$

Therefore:

$$
\frac{dy}{dx}
=
(2u)(3)
$$

Since:

$$
u=3x
$$

we get:

$$
\frac{dy}{dx}=18x
$$

---

# Why Do We Need the Chain Rule in Neural Networks?

Consider a very simple neuron:

$$
z=wx+b
$$

Then an activation function:

$$
a=f(z)
$$

Then a loss:

$$
L=L(a)
$$

The relationship is:

We want to know:

> How does changing the weight $w$ affect the loss $L$?

In mathematical form:

$$
\frac{\partial L}{\partial w}
$$

But there are intermediate variables between $w$ and $L$.

So we use the Chain Rule:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\times
\frac{\partial a}{\partial z}
\times
\frac{\partial z}{\partial w}
}
$$

This is one of the most important equations for understanding backpropagation.

---

# Breaking It Down

We have:

### 1. Loss with respect to activation

$$
\frac{\partial L}{\partial a}
$$

This tells us how the loss changes when the activation changes.

---

### 2. Activation with respect to weighted sum

$$
\frac{\partial a}{\partial z}
$$

This depends on the activation function.

For example, if:

$$
a=\sigma(z)
$$

then:

$$
\frac{\partial a}{\partial z}
=
\sigma(z)(1-\sigma(z))
$$

---

### 3. Weighted sum with respect to weight

Since:

$$
z=wx+b
$$

we have:

$$
\frac{\partial z}{\partial w}=x
$$

---

# Putting Everything Together

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\times
\frac{\partial a}{\partial z}
\times
x
}
$$

This gives us the gradient for the weight.

Then the optimizer can update the weight:

$$
w_{new}
=
w_{old}
-
\eta
\frac{\partial L}{\partial w}
$$

---

# The Big Picture

The Chain Rule allows the neural network to work backward through every operation:


At each step, we calculate a derivative and multiply it with the derivatives from the previous steps.

Overall Gradient

      ↑

Derivative 1 × Derivative 2 × Derivative 3


This is why it is called backpropagation:

> The effect of the loss is propagated backward through the network.

---

# Chain Rule in One Line

For:

$$
w \rightarrow z \rightarrow a \rightarrow L
$$

we calculate:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\frac{\partial a}{\partial z}
\frac{\partial z}{\partial w}
}
$$

---

# Key Takeaway

> The Chain Rule allows us to calculate how a weight affects the final loss when there are multiple intermediate steps.

Without the Chain Rule, calculating gradients through multiple neural-network layers would be extremely difficult.


Forward:

Weight → Weighted Sum → Activation → Prediction → Loss

Backward:

Loss → Activation → Weighted Sum → Weight

                ↓

           Chain Rule

                ↓

             Gradient


# 4. Backpropagation Through a Single Neuron

## Why Are We Doing This?

We already know:

- Forward propagation
- Loss function
- Gradient
- Chain Rule

Now we want to see how these concepts work together.

Consider a very simple neuron:

During backpropagation, we move in the opposite direction:

## Step 1 — Forward Propagation

Consider:

$$
x=2
$$

$$
w=0.5
$$

$$
b=0
$$

The weighted sum is:

$$
z=wx+b
$$

Therefore:

$$
z=(0.5)(2)+0
$$

$$
\boxed{z=1}
$$

Suppose we use the sigmoid activation function:

$$
a=\sigma(z)
$$

The sigmoid function is:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

For:

$$
z=1
$$

we get approximately:

$$
a=0.731
$$

So the prediction is:

$$
\boxed{\hat{y}=0.731}
$$

---

# Step 2 — Compare With the Actual Value

Suppose the actual value is:

$$
y=1
$$

The model predicted:

$$
\hat{y}=0.731
$$

So the prediction is not perfect.

We now calculate the loss.

For simplicity, let's use Mean Squared Error for this example:

$$
L=\frac{1}{2}(y-\hat{y})^2
$$

Therefore:

$$
L=\frac{1}{2}(1-0.731)^2
$$

$$
L\approx0.036
$$

So:

$$
\boxed{L\approx0.036}
$$

---

# Step 3 — Start Backpropagation

We now want to know:

> How much does the loss change if we change the weight $w$?

We need:

$$
\frac{\partial L}{\partial w}
$$

But the weight does not directly connect to the loss.

The relationship is:

Therefore, we use the Chain Rule:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\times
\frac{\partial a}{\partial z}
\times
\frac{\partial z}{\partial w}
}
$$

---

# Step 4 — Calculate Each Derivative

## Part 1: Loss With Respect to Prediction

Our loss is:

$$
L=\frac{1}{2}(y-a)^2
$$

Therefore:

$$
\frac{\partial L}{\partial a}
=
a-y
$$

Using:

$$
a=0.731
$$

and:

$$
y=1
$$

we get:

$$
\frac{\partial L}{\partial a}
=
0.731-1
$$

$$
\boxed{
\frac{\partial L}{\partial a}=-0.269
}
$$

---

## Part 2: Activation With Respect to Weighted Sum

For sigmoid:

$$
a=\sigma(z)
$$

Its derivative is:

$$
\boxed{
\frac{\partial a}{\partial z}
=
a(1-a)
}
$$

Therefore:

$$
\frac{\partial a}{\partial z}
=
(0.731)(1-0.731)
$$

$$
\boxed{
\frac{\partial a}{\partial z}\approx0.197
}
$$

---

## Part 3: Weighted Sum With Respect to Weight

We have:

$$
z=wx+b
$$

Therefore:

$$
\frac{\partial z}{\partial w}=x
$$

Since:

$$
x=2
$$

we get:

$$
\boxed{
\frac{\partial z}{\partial w}=2
}
$$

---

# Step 5 — Multiply the Derivatives

Now combine everything:

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\times
\frac{\partial a}{\partial z}
\times
\frac{\partial z}{\partial w}
$$

Substituting:

$$
\frac{\partial L}{\partial w}
=
(-0.269)(0.197)(2)
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w}\approx-0.106
}
$$

This is the gradient of the loss with respect to the weight.

---

# Step 6 — Update the Weight

Now the optimizer can use the gradient.

Suppose:

$$
\eta=0.1
$$

The update rule is:

$$
w_{new}
=
w_{old}
-
\eta
\frac{\partial L}{\partial w}
$$

Substituting:

$$
w_{new}
=
0.5-(0.1)(-0.106)
$$

Therefore:

$$
w_{new}
=
0.5+0.0106
$$

$$
\boxed{
w_{new}=0.5106
}
$$

The weight increased because the gradient was negative.

---

# Complete Process

We can now see the entire learning process:

# The Most Important Idea

The neural network does not randomly change the weight.

It follows a chain:

$$
\boxed{
Loss
\rightarrow
Gradients
\rightarrow
Weight\ Update
}
$$

And the Chain Rule tells us how to calculate the gradient:

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\times
\frac{\partial a}{\partial z}
\times
\frac{\partial z}{\partial w}
}
$$

This is the basic mathematical idea behind backpropagation.

---
# Python Function

We can implement this simple example manually:

```python
import math


def sigmoid(z):
    return 1 / (1 + math.exp(-z))


def backpropagation(x, y, w, b, learning_rate):

    # Forward propagation
    z = w * x + b
    a = sigmoid(z)

    # Loss
    loss = 0.5 * (y - a) ** 2

    # Backpropagation
    dL_da = a - y
    da_dz = a * (1 - a)
    dz_dw = x

    # Gradient
    dL_dw = dL_da * da_dz * dz_dw

    # Weight update
    new_w = w - learning_rate * dL_dw

    return new_w, loss, dL_dw
```

```python
# Example

new_w, loss, gradient = backpropagation(
    x=2,
    y=1,
    w=0.5,
    b=0,
    learning_rate=0.1
)

print("Loss:", loss)
print("Gradient:", gradient)
print("Updated weight:", new_w)
```

---

## Key Takeaway

> Backpropagation uses the Chain Rule to calculate how much the loss depends on each parameter.

For a single neuron:

Backpropagation travels backward:

And after calculating the gradient, the optimizer updates the weight.

This is the core mechanism by which a neural network learns.

# 5. Backpropagation Through Multiple Layers

## Why Do We Need This?

We just calculated the gradient for a single neuron:

$$
\frac{\partial L}{\partial w}
$$

But a neural network can contain:

There may be many weights:

$$
w_1,w_2,w_3,\ldots,w_n
$$

We need to calculate the gradient for every weight.

Backpropagation does this by applying the Chain Rule repeatedly.

---

## Simple Two-Layer Network

Consider:

The forward path is:

$$
x
\rightarrow
z_1
\rightarrow
a_1
\rightarrow
z_2
\rightarrow
a_2
\rightarrow
L
$$

Where:

- $x$ = Input
- $z_1$ = Weighted sum in hidden layer
- $a_1$ = Hidden-layer activation
- $z_2$ = Weighted sum in output layer
- $a_2$ = Prediction
- $L$ = Loss

---

# Forward Propagation

The hidden neuron calculates:

$$
z_1=w_1x+b_1
$$

Then:

$$
a_1=f(z_1)
$$

The output neuron receives $a_1$:

$$
z_2=w_2a_1+b_2
$$

Then:

$$
a_2=f(z_2)
$$

Finally:

$$
L=L(a_2,y)
$$

So:


---

# Backpropagation

Now we travel in the opposite direction:

We want to calculate:

$$
\frac{\partial L}{\partial w_2}
$$

and:

$$
\frac{\partial L}{\partial w_1}
$$

---

# Gradient for the Output Weight

For $w_2$, the path is:

$$
w_2
\rightarrow
z_2
\rightarrow
a_2
\rightarrow
L
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w_2}
=
\frac{\partial L}{\partial a_2}
\times
\frac{\partial a_2}{\partial z_2}
\times
\frac{\partial z_2}{\partial w_2}
}
$$

This is the same Chain Rule idea we used for the single neuron.

Since:

$$
z_2=w_2a_1+b_2
$$

we have:

$$
\frac{\partial z_2}{\partial w_2}=a_1
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w_2}
=
\frac{\partial L}{\partial a_2}
\times
\frac{\partial a_2}{\partial z_2}
\times
a_1
}
$$

---

# Gradient for the Hidden-Layer Weight

Now consider $w_1$.

Its effect on the loss travels through more intermediate variables:

$$
w_1
\rightarrow
z_1
\rightarrow
a_1
\rightarrow
z_2
\rightarrow
a_2
\rightarrow
L
$$

Therefore:

$$
\boxed{
\frac{\partial L}{\partial w_1}
=
\frac{\partial L}{\partial a_2}
\times
\frac{\partial a_2}{\partial z_2}
\times
\frac{\partial z_2}{\partial a_1}
\times
\frac{\partial a_1}{\partial z_1}
\times
\frac{\partial z_1}{\partial w_1}
}
$$

Notice something important:

> The deeper we go backward, the more Chain Rule terms we encounter.

---

# Why Is It Called "Backpropagation"?

Because the information about the loss is propagated backward through the network.

The network uses this backward calculation to determine how each parameter contributed to the final error.

---

# What Happens After the Gradients Are Calculated?

Once we have:

$$
\frac{\partial L}{\partial w_1}
$$

and:

$$
\frac{\partial L}{\partial w_2}
$$

we can update the weights.

For example:

$$
w_1^{new}
=
w_1^{old}
-
\eta
\frac{\partial L}{\partial w_1}
$$

and:

$$
w_2^{new}
=
w_2^{old}
-
\eta
\frac{\partial L}{\partial w_2}
$$

The same process is performed for all weights and biases.

---

# The Complete Learning Process

              FORWARD
                 ↓
        Input → Hidden → Output → Prediction
                              ↓
                            Loss
                              ↓
                        ─────────────┘
                           BACKWARD
                              ↓
                       Calculate Gradients
                              ↓
                       Update Parameters
                              ↓
                         Repeat Again



This process happens repeatedly during training.

---

# Important Point

Backpropagation does not mean that the model simply sends the prediction backward.

Instead, it propagates the effect of the loss backward through the computational steps.

The goal is to calculate:

$$
\boxed{
\frac{\partial L}{\partial w_i}
}
$$

for every parameter $w_i$.

---

# Key Takeaway

> In a multi-layer neural network, backpropagation applies the Chain Rule repeatedly to calculate the gradient of the loss with respect to every weight and bias.

The overall process is:

$$
\boxed{
Forward
\rightarrow
Loss
\rightarrow
Backpropagation
\rightarrow
Gradients
\rightarrow
Parameter\ Update
}
$$

$$

- $\frac{\partial L}{\partial w}$ = Gradient
- $\eta$ = Learning rate

}
$$

> How does the loss change if this weight changes?

\boxed{

> How should we change the weight to reduce the loss?

# 7. Learning Rate in Gradient Descent

## What is Learning Rate?

The learning rate controls how large a step the model takes when updating its weights.

It is represented by:

$$
\eta
$$

The basic update equation is:

$$
\boxed{
w_{new}
=
w_{old}
-
\eta
\frac{\partial L}{\partial w}
}
$$

Where:

- $w$ = Weight
- $\eta$ = Learning rate
- $\frac{\partial L}{\partial w}$ = Gradient
- $L$ = Loss

---

## Why Do We Need a Learning Rate?

Backpropagation tells us the direction and magnitude of the gradient.

But we still need to decide:

> How big a step should we take in that direction?

The learning rate controls this step.

---

# Small Learning Rate

Suppose:

$$
\eta=0.01
$$

The model makes small updates.

### Advantage

The learning process is more gradual and can make fine adjustments.

### Disadvantage

Learning can become very slow.

The model may require many iterations to reach a low-loss region.

---

# Large Learning Rate

Suppose:

$$
\eta=1
$$

The model makes much larger updates.

### Advantage

The model can move quickly.

### Disadvantage

The updates can become too large.

The model may:

- Overshoot the minimum
- Oscillate around the minimum
- Fail to converge

---

# Numerical Example

Suppose:

$$
w=2
$$

and:

$$
\frac{\partial L}{\partial w}=0.5
$$

---

## Learning Rate = 0.1

$$
w_{new}
=
2-(0.1)(0.5)
$$

$$
w_{new}=1.95
$$

So:

$$
2\rightarrow1.95
$$

The update is:

$$
\Delta w=-0.05
$$

---

## Learning Rate = 0.01

$$
w_{new}
=
2-(0.01)(0.5)
$$

$$
w_{new}=1.995
$$

The update is much smaller:

$$
\Delta w=-0.005
$$

---

## Learning Rate = 1

$$
w_{new}
=
2-(1)(0.5)
$$

$$
w_{new}=1.5
$$

Now the update is:

$$
\Delta w=-0.5
$$

So the learning rate directly controls the size of the parameter update.

---

# Learning Rate and the Loss Landscape

Imagine the loss as a valley:

The goal of Gradient Descent is to move toward the low-loss region.

### Learning rate too small

● → ● → ● → ● → ● → ●


Very slow movement.

### Suitable learning rate

●   →   ●   →   ●   →   ●

              ↓

          Minimum


The model gradually approaches the low-loss region.

### Learning rate too large

The model can jump across the minimum instead of settling near it.

---

# Important Point

The learning rate does not determine the direction of the update.

The gradient determines the direction.

The learning rate determines the size of the step.

---

# Python Function

We can represent the Gradient Descent update with a simple function:


```python
def gradient_descent_update(weight, gradient, learning_rate):
    
    new_weight = weight - learning_rate * gradient
    
    return new_weight
```

```python
# Example

weight = 2
gradient = 0.5
learning_rate = 0.1

new_weight = gradient_descent_update(
    weight,
    gradient,
    learning_rate
)

print("Updated weight:", new_weight)
```

# Learning Rate in a Neural Network

A neural network has many parameters:

$$
w_1,w_2,w_3,\ldots,w_n
$$

Each parameter gets its own gradient.

But the same learning rate can be used in the basic Gradient Descent update:

$$
w_i^{new}
=
w_i^{old}
-
\eta
\frac{\partial L}{\partial w_i}
$$

So:


---

# Key Takeaway

> The learning rate controls the size of each parameter update during Gradient Descent.

Remember:

$$
\boxed{
\text{Gradient} = \text{direction}
}
$$
$$
\boxed{
\text{Learning Rate} = \text{step size}
}
$$

And:

$$
\boxed{
w_{new}
=
w_{old}
-
\eta
\frac{\partial L}{\partial w}
}
$$

A good learning rate allows the model to move toward lower loss efficiently without making excessively large or unnecessarily small steps.
