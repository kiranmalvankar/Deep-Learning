## 1. What is LSTM?

Long Short-Term Memory (LSTM) is a specialized type of recurrent neural network designed to learn and retain important information over longer sequences.

LSTM was introduced to address limitations of traditional RNNs, especially the vanishing-gradient problem that can make learning long-term dependencies difficult.

In simple words:

> LSTM is an improved form of RNN that uses a memory cell and gates to control what information should be remembered, updated, and forgotten.

# 8. Forget Gate

The forget gate decides which information from the previous cell state should be retained or discarded.

The equation is:

$$
\boxed{
f_t=
\sigma
(W_f[h_{t-1},x_t]+b_f)
}
$$

Where:

- $f_t$ = Forget-gate output
- $\sigma$ = Sigmoid function
- $h_{t-1}$ = Previous hidden state
- $x_t$ = Current input
- $W_f$ = Forget-gate weights
- $b_f$ = Forget-gate bias

The sigmoid produces values between:

$$
0 \leq f_t \leq 1
$$

# 10. Input Gate

# 11. Candidate Cell State

LSTM also creates candidate information that could be added to memory.

$$
\boxed{
\tilde{C}_t=
\tanh
(W_C[h_{t-1},x_t]+b_C)
}
$$

The tanh function produces values approximately between:

$$
-1 \text{ and } 1
$$

This candidate represents new information that could be stored.

---

# 12. Updating the Cell State

The new cell state is:

$$
\boxed{
C_t=
f_t\odot C_{t-1}
+
i_t\odot\tilde{C}_t
}
$$

Where:

- $C_{t-1}$ = Previous cell state
- $f_t$ = Forget gate
- $i_t$ = Input gate
- $\tilde{C}_t$ = Candidate cell state
- $\odot$ = Element-wise multiplication

This equation is the heart of LSTM memory management.

# 18. Why Does This Help With Long-Term Dependencies?

The cell state provides a pathway through which information can be carried across many time steps.

Conceptually:

The gates can control how information is retained or modified.

This can make it easier for useful information to persist over longer sequences than in a basic RNN.

---

# 19. LSTM and Vanishing Gradients

Traditional RNNs can suffer from vanishing gradients because gradients may shrink as they propagate through many time steps.

LSTM's gated cell-state mechanism provides a more effective pathway for information and gradient flow.

Conceptually:


LSTM does not completely eliminate all optimization difficulties, but it was specifically designed to make long-term dependencies easier to learn.

# 34. LSTM Advantages


1. Better at learning long-term dependencies
   than traditional RNNs

2. Uses explicit memory through the cell state

3. Gates control information flow

4. Can process variable-length sequences

5. Useful for text, time series and biological sequences

6. Can be applied to SMILES-based molecular modeling


---

# 35. LSTM Limitations

1. More parameters than a basic RNN

2. More computationally expensive

3. More complex architecture

4. Sequential processing limits parallelization

5. Can still be difficult to train on very long sequences

6. Transformers often outperform recurrent models
   for many modern sequence tasks
