# 1. What is an Optimizer?

## What is an Optimizer?

An optimizer is an algorithm that updates the model's parameters (weights and biases) during training so that the model's loss decreases.

In simple terms:

> The optimizer decides how the model should change its weights to improve its predictions.

---

## Where Does the Optimizer Come in the Training Workflow?

We already learned the basic training process:

  ↓
Update Weights
  ↓
Next Iteration


The optimizer uses the gradients calculated by backpropagation to update the weights.

---

Suppose our model makes a prediction:

# Basic Weight Update

The simplest gradient-descent update is:

$$
w_{new}=w_{old}-\eta\frac{\partial L}{\partial w}
$$

Where:

- $w$ = weight
- $L$ = loss
- $\frac{\partial L}{\partial w}$ = gradient
- $\eta$ = learning rate

The optimizer is responsible for determining how these parameter updates are performed.

---

# Learning Rate

The learning rate controls the size of the update.

# What Happens if Learning Rate Is Too Small?

# 3. Momentum

## What is Momentum?

Momentum is an optimization technique that improves Gradient Descent by using information from previous updates to influence the current update.

Instead of looking only at the current gradient, Momentum keeps a kind of memory of previous gradients.

The basic idea is:

---

# Why Do We Need Momentum?

Basic Gradient Descent can sometimes behave inefficiently.

For example, imagine the loss surface looks like a narrow valley:

The gradient may cause the model to move:

← → ← → ← →


instead of moving smoothly toward the minimum.

This can make optimization slower.

Momentum helps by maintaining movement in a useful direction.

---

# Simple Analogy

Imagine pushing a heavy ball down a hill.

Without momentum:

The previous movement influences the current movement.

---

# Basic Gradient Descent

Without Momentum:

$$
w_{new}
=
w_{old}
-
\eta g
$$

where:

- $w$ = weight
- $\eta$ = learning rate
- $g$ = current gradient

The update depends only on the current gradient.

---

# Gradient Descent With Momentum

Momentum introduces a velocity term.

A simplified form is:

$$
v_t
=
\beta v_{t-1}
+
g_t
$$

Then:

$$
w_t
=
w_{t-1}
-
\eta v_t
$$

Where:

- $v_t$ = current velocity/update direction
- $v_{t-1}$ = previous velocity
- $g_t$ = current gradient
- $\beta$ = momentum coefficient
- $\eta$ = learning rate

---

# What Does β Mean?

$\beta$ controls how much influence the previous update has.

For example:


β = 0


means essentially no momentum contribution.

As $\beta$ increases:

A commonly used value is around:

β = 0.9


although the appropriate value depends on the problem and optimizer.

---

$$

$$

The optimizer now uses this accumulated direction when updating the weight.

The important idea is not the exact number.

The important idea is:

> The current update is influenced by previous updates.

---

# Why Can Momentum Be Faster?

Suppose gradients repeatedly point in approximately the same direction:

Momentum accumulates this information:

So movement in the consistent direction can become faster.

---

# Momentum Helps With Oscillation

Suppose the gradient alternates strongly in one direction:


Basic Gradient Descent may oscillate.

Momentum can reduce this kind of inefficient movement while maintaining progress in the useful direction.

Conceptually:


The exact behavior depends on the loss surface and hyperparameters, but this is the main intuition.

# 5. Adam Optimizer

## What is Adam?

Adam (Adaptive Moment Estimation) is an optimization algorithm used to update the weights and biases of a neural network during training.

Adam combines two important ideas:

In simple words:

> Adam decides how much each parameter should be updated by considering both the direction and the recent magnitude of its gradients.

---

# Why Do We Need Adam?

During neural-network training, we need to continuously update the weights.

We already know:

The `??? is where the **optimizer** comes in.

A basic Gradient Descent update is:

$$
w_{new}=w_{old}-\eta g
$$

The problem is that a simple update may not always be efficient.

Different parameters can have:

- Different gradient magnitudes
- Different gradient directions
- Noisy gradients
- Very small or very large gradients

Adam tries to handle these situations more effectively.

---

# Main Idea Behind Adam

Adam maintains two types of information about the gradients.

## 1. First Moment — Direction

Adam keeps a moving average of the gradients.

It is represented by:

$$
m_t
$$

This tells Adam about the **general direction in which the gradients have been moving**.

This is similar to the idea of **Momentum**.



---

## 2. Second Moment — Magnitude

Adam also keeps a moving average of the **squared gradients**.

It is represented by:

$$
v_t
$$

This gives information about the recent magnitude of the gradients.

This is related to the idea behind **RMSProp**.



---

# Adam Conceptually


So the easiest way to remember Adam is:

$$
\boxed{
Adam = Direction + Magnitude
}
$$

---

# Adam Equations

## Step 1 — Calculate First Moment

$$
m_t =
\beta_1m_{t-1}
+
(1-\beta_1)g_t
$$

Where:

- $m_t$ = Current first moment
- $m_{t-1}$ = Previous first moment
- $g_t$ = Current gradient
- $\beta_1$ = First-moment decay factor

---

## Step 2 — Calculate Second Moment

$$
v_t =
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2
$$

Where:

- $v_t$ = Current second moment
- $v_{t-1}$ = Previous second moment
- $g_t^2$ = Squared current gradient
- $\beta_2$ = Second-moment decay factor

---

# Why Does Adam Need Bias Correction?

At the beginning of training:

$$
m_0=0
$$

and:

$$
v_0=0
$$

Because both moving averages start at zero, they can be biased toward zero during the initial updates.

Adabias correction **bias correction**.

### Corrected first moment

$$
\hat{m}_t=
\frac{m_t}{1-\beta_1^t}
$$

### Corrected second moment

$$
\hat{v}_t=
\frac{v_t}{1-\beta_2^t}
$$

---

# Final Adam Update

After calculating the corrected moments:

$$
\boxed{
w_t =
w_{t-1}
-
\eta
\frac{\hat{m}_t}
{\sqrt{\hat{v}_t}+\epsilon}
}
$$

Where:

- $w_t$ = New weight
- $w_{t-1}$ = Previous weight
- $\eta$ = Learning rate
- $\hat{m}_t$ = Bias-corrected first moment
- $\hat{v}_t$ = Bias-corrected second moment
- $\epsilon$ = Small value added for numerical stability

---

# What Are β₁ and β₂?

Adam has two important decay parameters.

### β₁

Controls the moving average of gradients.

A commonly used default is:

$$
\beta_1=0.9
$$

---

### β₂

Controls the moving average of squared gradients.

A commonly used default is:

$$
\beta_2=0.999
$$

---

# Common Adam Defaults

A commonly used configuration is:

These are defaults, not universal rules.

They can be changed depending on the problem.
