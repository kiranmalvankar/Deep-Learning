# 14. Artificial Neural Network (ANN)

# 1. What is an ANN?

## Definition

Artificial Neural Network (ANN) is a machine-learning model inspired by the basic information-processing structure of biological neural networks.

An ANN consists of interconnected computational units called neurons, organized into layers.

A basic ANN contains:

Input Layer
     ↓
Hidden Layer(s)
     ↓
Output Layer


The neurons are connected through weights, and each neuron can have a bias.

The network learns appropriate weights and biases from training data.

# 9. Neuron Calculation

A neuron first calculates a weighted sum.

For inputs:

$$
x_1,x_2,...,x_n
$$

with weights:

$$
w_1,w_2,...,w_n
$$

the neuron calculates:

$$
z =
\sum_{i=1}^{n}w_ix_i+b
$$

or:

$$
\boxed{
z=w_1x_1+w_2x_2+\cdots+w_nx_n+b
}
$$

Then an activation function is applied:

$$
a=f(z)
$$

# 13. Matrix Representation

For a dense layer:

$$
Z=XW+b
$$

Then:

$$
A=f(Z)
$$

Where:

- $X$ = Input matrix
- $W$ = Weight matrix
- $b$ = Bias
- $Z$ = Weighted sum
- $f$ = Activation function
- $A$ = Activated output

This is the mathematical foundation of a dense neural network.

---

# 14. Forward Propagation in ANN

During forward propagation:

For each layer:

$$
Z^{(l)}=A^{(l-1)}W^{(l)}+b^{(l)}
$$

Then:

$$
A^{(l)}=f(Z^{(l)})
$$

This process continues until the output layer.

# 38. Limitations of ANN

ANNs also have limitations

1. Can require substantial training data

2. Can overfit

3. Hyperparameters need to be selected

4. Training can be computationally expensive

5. Often less interpretable than simpler models

6. Results can depend strongly on architecture
   and preprocessing
