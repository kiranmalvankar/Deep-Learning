# Loss function for Regression

# 01: MAE — Mean Absolute Error

MAE = Mean Absolute Error

It measures the average absolute difference between the actual values and predicted values.

In simple words:

MAE tells us, on average, how far the predictions are from the actual values.

## Mathematical Equation (MAE)

$$
MAE =
\frac{1}{n}
\sum_{i=1}^{n}
\left|y_i-\hat{y}_i\right|
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $n$ = Number of observations
- $|y_i-\hat{y}_i|$ = Absolute erro

## Why "Absolute" in MAE?

When calculating error, the difference between the actual and predicted value can be either positive or negative.

For example:

### Example 1

Actual value:

$$
y = 10
$$

Predicted value:

$$
\hat{y} = 8
$$

Error:

$$
y-\hat{y}=10-8=2
$$

### Example 2

Actual value:

$$
y = 10
$$

Predicted value:

$$
\hat{y} = 12
$$

Error:

$$
y-\hat{y}=10-12=-2
$$

If we simply average these errors:

$$
\frac{2+(-2)}{2}=0
$$

This would incorrectly suggest that there is no error.

Therefore, MAE uses the absolute value of the error:

$$
|2|=2
$$

and

$$
|-2|=2
$$

The negative sign is removed, so both errors represent a distance of 2 from the actual value.

### Therefore:

> "Absolute" means we consider only the magnitude of the error, regardless of whether the prediction is above or below the actual value.

The MAE equation is:

$$
MAE =
\frac{1}{n}
\sum_{i=1}^{n}
\left|y_i-\hat{y}_i\right|
$$

```python
# function

# Absolute error: 

def absolute_error(actual, predicted):
    return abs(actual - predicted)

# MAE Function:

def mae(actual, predicted):
    errors = abs(actual - predicted)
    return errors.mean()
```

# 02: MSE — Mean Squared Error

MSE squares each error before taking the average.

# Mathematical Equation


$$
MSE =
\frac{1}{n}
\sum_{i=1}^{n}
\left(y_i-\hat{y}_i\right)^2
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $n$ = Number of observations
- $\left(y_i-\hat{y}_i\right)^2$ = Squared error

## Understanding MSE in Simple Terms

Suppose our model makes three predictions:

| Actual | Prediction |
|---:|---:|
| 10 | 8 |
| 10 | 9 |
| 10 | 15 |

### Step 1 — Calculate the errors

$$
Error = Actual - Prediction
$$

| Actual | Prediction | Error |
|---:|---:|---:|
| 10 | 8 | 2 |
| 10 | 9 | 1 |
| 10 | 15 | -5 |

The errors can be positive or negative.

If we simply average them:

$$
\frac{2+1+(-5)}{3}=-0.67
$$

This does not properly represent the amount of error because positive and negative errors can cancel each other.

---

### Step 2 — Square every error

MSE solves this problem by squaring each error:

| Error | Squared Error |
|---:|---:|
| 2 | $2^2=4$ |
| 1 | $1^2=1$ |
| -5 | $(-5)^2=25$ |

Squaring makes all errors positive.

---

### Step 3 — Take the average

$$
MSE=\frac{4+1+25}{3}
$$

$$
\boxed{MSE=10}
$$

Therefore:

> MSE = Square the errors → Then take their average.

---

## Why Do We Square the Errors?

The most important reason is that squaring gives a much larger penalty to large errors.

For example:

| Error | MAE contribution | MSE contribution |
|---:|---:|---:|
| 2 | 2 | $2^2=4$ |
| 10 | 10 | $10^2=100$ |

With MAE:

$$
2 \rightarrow 2
$$

$$
10 \rightarrow 10
$$

With MSE:

$$
2^2=4
$$

$$
10^2=100
$$

Therefore, MSE strongly penalizes large errors.

> A large prediction error has a much greater effect on MSE than a small prediction error.

This makes MSE more sensitive to large errors and outliers.

---

## Simple Way to Remember

### MAE

> "How far am I wrong?"

### MSE

> "How far am I wrong, while giving much more penalty to large mistakes?"

The mathematical definition is:

$$
\boxed{
MSE=
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

🎯 Why would we choose one over the other?

MAE

Useful when you want the model to be less dominated by very large errors/outliers.

MSE

Useful when you want the model to strongly punish large errors.

For example, if a few predictions are extremely far from the actual values, MSE will put much more pressure on the model to reduce those errors.
    
One-line difference
    
MAE measures error linearly; MSE squares the error, making large errors disproportionately important and producing larger gradients during training.


```python
# Python Function

import numpy as np

def mse(actual, predicted):
    errors = actual - predicted
    squared_errors = errors ** 2
    return squared_errors.mean()

# Example

actual = np.array([10, 20, 30])
predicted = np.array([8, 23, 27])

mse_value = mse(actual, predicted)

print("MSE:", mse_value)
```

# 03_RMSE — Root Mean Squared Error

We have learned:

MAE → absolute error

MSE → squared error

Now RMSE is very straightforward.

1. What is RMSE?
   
RMSE = Root Mean Squared Error

RMSE is simply the square root of MSE.

RMSE measures the typical size of prediction errors while giving more importance to larger errors.

## Mathematical Equation (RMSE)

RMSE stands for Root Mean Squared Error.

It is calculated by taking the square root of the Mean Squared Error (MSE).

$$
RMSE = \sqrt{MSE}
$$

Therefore:

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

Where:

- $y_i$ = Actual value
- $\hat{y}_i$ = Predicted value
- $n$ = Number of observations

### Simple Process

$$
\text{Error}
\rightarrow
\text{Square}
\rightarrow
\text{Average}
\rightarrow
\text{Square Root}
\rightarrow
RMSE
$$

## Why Do We Need the Square Root in RMSE?

MSE squares the errors before averaging them.

For example:

$$
Error = 2
$$

$$
2^2 = 4
$$

If the original target is measured in a particular unit, squaring the error also squares the unit.

For example:

$$
\text{Target unit} = U
$$

Then:

$$
\text{MSE unit} = U^2
$$

This makes MSE less intuitive to interpret because it is expressed in squared units.

RMSE solves this by taking the square root of MSE:

$$
RMSE = \sqrt{MSE}
$$

For example:

$$
MSE = 9
$$

Then:

$$
RMSE = \sqrt{9}=3
$$

Therefore, RMSE returns the error measurement to the same units as the original target.

### Simple Process

$$
\text{Error}
\rightarrow
\text{Square}
\rightarrow
\text{Average}
\rightarrow
\text{Square Root}
\rightarrow
RMSE
$$

### Key Idea

> The square root brings the result back to the original scale of the target while retaining the effect of squaring the errors.

Therefore:

- MSE → squared units
- RMSE → original units

```python
# Python Function

import numpy as np

def rmse(actual, predicted):
    errors = actual - predicted
    squared_errors = errors ** 2
    mse_value = np.mean(squared_errors)
    return np.sqrt(mse_value)


# Example

actual = np.array([10, 20, 30])
predicted = np.array([8, 23, 27])

rmse_value = rmse(actual, predicted)

print("RMSE:", rmse_value)
```

# MAE vs MSE vs RMSE

We have three commonly used regression error metrics:

- MAE — Mean Absolute Error
- MSE — Mean Squared Error
- RMSE — Root Mean Squared Error

All three measure the difference between actual and predicted values, but they treat errors differently.

---

## 1. MAE

MAE calculates the absolute error and then takes the average.

$$
MAE =
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
$$

### Main characteristic

MAE treats errors approximately linearly.

For example:

$$
1 \rightarrow 1
$$

$$
5 \rightarrow 5
$$

$$
10 \rightarrow 10
$$

A large error is larger, but it is not disproportionately amplified.

### Key idea

> MAE tells us the average size of the prediction error and is relatively less sensitive to outliers.

---

## 2. MSE

MSE squares each error before taking the average.

$$
MSE =
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
$$

### Main characteristic

Squaring makes large errors much more influential.

For example:

$$
1^2=1
$$

$$
5^2=25
$$

$$
10^2=100
$$

### Key idea

> MSE strongly penalizes large prediction errors and is therefore more sensitive to outliers.

---

## 3. RMSE

RMSE is the square root of MSE.

$$
RMSE =
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

It retains the effect of squaring errors but brings the final value back to the original units of the target.

### Key idea

> RMSE gives more influence to large errors while remaining in the original units of the target.

---

## 4. Simple Comparison

Suppose the errors are:

$$
1,\;2,\;10
$$

### MAE

$$
MAE=\frac{1+2+10}{3}=4.33
$$

### MSE

$$
MSE=\frac{1^2+2^2+10^2}{3}
$$

$$
MSE=\frac{1+4+100}{3}=35
$$

### RMSE

$$
RMSE=\sqrt{35}\approx5.92
$$

Notice that the large error of 10 has a much stronger effect on MSE and RMSE.

---

## 5. Comparison Table

| Metric | Error treatment | Sensitivity to large errors | Units |
|---|---|---|---|
| MAE | Absolute error | Lower | Same as target |
| MSE | Squared error | High | Squared units |
| RMSE | Squared error + square root | High | Same as target |

---

## 6. When Can We Prefer Each One?

### MAE

Useful when:

- We want an easily interpretable average error.
- We don't want very large errors to dominate the metric.
- The dataset may contain outliers.

### MSE

Useful when:

- Large errors should be strongly penalized.
- We want a smooth loss function that is convenient for gradient-based optimization.
- Large mistakes are particularly undesirable.

### RMSE

Useful when:

- Large errors should receive more importance.
- We want the final metric in the same units as the target.
- We want an interpretable metric while retaining MSE's sensitivity to large errors.

---

## 7. Important Point

There is no universally "best" metric.

The appropriate choice depends on:

- The problem
- The distribution of errors
- The presence of outliers
- How costly large errors are
- Whether the metric is being used for training or evaluation

For example, a model can be evaluated using both MAE and RMSE to understand different aspects of its prediction errors.

---

## Simple Memory Trick

MAE → Average distance

MSE → Large errors are strongly penalized

RMSE → MSE behavior + original units

# Loss Function for Classification

# 01_Binary Cross-Entropy (BCE)

Binary Cross-Entropy (BCE) is a loss function commonly used for binary classification.

It measures how different the model's predicted probability is from the actual class.

BCE gives a small loss when the model is confidently correct and a large loss when the model is confidently wrong.

## Why Probability in Binary Classification?

In binary classification, the model needs to choose between two classes.

For example:

- $0$ → Inactive
- $1$ → Active

Instead of immediately producing only 0 or 1, the neural network first produces a probability between 0 and 1.

For example:

$$
0.90 = 90\% \text{ probability of class 1}
$$

$$
0.20 = 20\% \text{ probability of class 1}
$$

This probability tells us not only which class the model predicts, but also how confident the model is.

---

## How Does It Work?

The output neuron first calculates a weighted sum:

$$
z = Wx+b
$$

The value $z$ can be any number:

$$
-\infty < z < +\infty
$$

For example:

$$
z=2.2
$$

But we need a value between 0 and 1 to represent probability.

Therefore, we apply the Sigmoid activation function:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

So:

$$
z
\rightarrow
\text{Sigmoid}
\rightarrow
\hat{y}
$$

where $\hat{y}$ is the predicted probability.

---

## Example

Suppose the output neuron produces:

$$
z=2.2
$$

After applying Sigmoid:

$$
\hat{y}=\frac{1}{1+e^{-2.2}}
$$

Approximately:

$$
\hat{y}=0.90
$$

Therefore:

$$
0.90=90\%
$$

The model predicts a 90% probability of class 1.

---

## From Probability to Class

We can convert the probability into a final class using a threshold.

A common threshold is $0.5$:

| Predicted Probability | Predicted Class |
|---:|---|
| $<0.5$ | Class 0 |
| $\geq0.5$ | Class 1 |

For example:

$$
0.90 \rightarrow Class\ 1
$$

$$
0.20 \rightarrow Class\ 0
$$

The important point is that the probability is produced first. The class label is obtained afterward using a decision threshold.

---

## Why Not Directly Predict 0 or 1?

Suppose the actual class is:

$$
y=1
$$

Compare two predictions:

### Model A

$$
\hat{y}=0.51
$$

### Model B

$$
\hat{y}=0.99
$$

Both would be classified as class 1 using a 0.5 threshold.

However, Model B is much more confident.

Therefore, using probabilities allows the loss function to distinguish between:

- barely correct predictions
- highly confident correct predictions
- highly confident incorrect predictions

This information is important for Binary Cross-Entropy.

---

## Complete Process

The binary classification process is:

$$
\text{Input}
\rightarrow
\text{Weighted Sum}
\rightarrow
z
\rightarrow
\text{Sigmoid}
\rightarrow
\text{Probability}
\rightarrow
\text{Threshold}
\rightarrow
\text{Class}
$$

For example:

$$
z=2.2
\rightarrow
\text{Sigmoid}
\rightarrow
0.90
\rightarrow
0.90>0.5
\rightarrow
Class\ 1
$$

### Key Idea

> The neural network first predicts a probability, which represents its confidence in the positive class. A threshold is then used to convert that probability into a final class label.

 # Mathematical Equation

## Binary Cross-Entropy (BCE)

For a single observation:

$$
BCE =
-\left[
y\log(\hat{y})
+
(1-y)\log(1-\hat{y})
\right]
$$

Where:

- $y$ = Actual class ($0$ or $1$)
- $\hat{y}$ = Predicted probability of class $1$

For $n$ observations:

$$
BCE =
-\frac{1}{n}
\sum_{i=1}^{n}
\left[
y_i\log(\hat{y}_i)
+
(1-y_i)\log(1-\hat{y}_i)
\right]
$$

## Why Does Binary Cross-Entropy Use Logarithm?

The logarithm is used in Binary Cross-Entropy (BCE) to make the loss:

- Very small for confident correct predictions
- Very large for confident incorrect predictions

For a single observation, BCE is:

$$
BCE =
-\left[
y\log(\hat{y})
+
(1-y)\log(1-\hat{y})
\right]
$$

---

## Case 1: Actual Class = 1

When:

$$
y=1
$$

The BCE equation becomes:

$$
BCE=-\log(\hat{y})
$$

Now consider different predicted probabilities:

| Predicted Probability | BCE Loss |
|---:|---:|
| 0.99 | $\approx 0.01$ |
| 0.90 | $\approx 0.11$ |
| 0.50 | $\approx 0.69$ |
| 0.10 | $\approx 2.30$ |
| 0.01 | $\approx 4.61$ |

When the actual class is 1:

$$
0.99 \rightarrow \text{Very small loss}
$$

$$
0.01 \rightarrow \text{Very large loss}
$$

Therefore, predicting a high probability for the correct class is rewarded, while predicting a very low probability for the correct class is strongly penalized.

---

## Why Does the Logarithm Create This Behavior?

The logarithm has an important property:

$$
\log(1)=0
$$

As the probability approaches zero:

$$
\log(0)\rightarrow-\infty
$$

Because BCE contains a negative sign:

$$
-\log(0)\rightarrow+\infty
$$

Therefore, if the actual class is 1 and the model predicts a probability very close to 0, the BCE loss becomes very large.

---

## Example

Suppose:

$$
Actual=1
$$

### Prediction = 0.99

The model is very confident and correct:

$$
BCE=-\log(0.99)\approx0.01
$$

So the loss is very small.

### Prediction = 0.01

The model is very confident but wrong:

$$
BCE=-\log(0.01)\approx4.61
$$

So the loss is much larger.

---

## Why Is Confidence Important?

Consider two predictions for an actual class of 1:

$$
0.51
$$

and

$$
0.99
$$

Both may be classified as class 1 using a threshold of 0.5.

However:

- $0.51$ → The model is only slightly confident.
- $0.99$ → The model is highly confident.

BCE preserves this probability information.

Similarly, predicting:

$$
0.01
$$

for an actual class of 1 means the model is highly confident in the wrong answer, so BCE gives it a very large penalty.

---

## Simple Intuition

Think of BCE as saying:

> "If the correct class is 1, give it a high probability. The more confidently you give it a low probability, the more severely I will penalize you."

Therefore:

$$
\text{Confident + Correct}
\rightarrow
\text{Small Loss}
$$

$$
\text{Uncertain}
\rightarrow
\text{Moderate Loss}
$$

$$
\text{Confident + Wrong}
\rightarrow
\text{Large Loss}
$$

---

## Key Takeaway

> The logarithm in BCE creates a loss function that strongly penalizes confident incorrect predictions while giving very small loss to confident correct predictions.

This is why BCE works well with probability-based binary classification.

## BCE When the Actual Class Is 0

So far, we considered the case:

$$
y=1
$$

Now let's consider:

$$
y=0
$$

For binary classification:

- $0$ → Negative class
- $1$ → Positive class

The BCE equation is:

$$
BCE =
-\left[
y\log(\hat{y})
+
(1-y)\log(1-\hat{y})
\right]
$$

When:

$$
y=0
$$

the equation becomes:

$$
BCE=-\log(1-\hat{y})
$$

Here, $\hat{y}$ is the predicted probability of class 1.

---

## How Does It Work?

Suppose the actual class is:

$$
y=0
$$

The model should therefore give a low probability to class 1.

### Prediction = 0.01

The model predicts only a 1% probability of class 1.

$$
BCE=-\log(1-0.01)
$$

$$
BCE=-\log(0.99)
$$

$$
BCE\approx0.01
$$

So the loss is very small.

The model is confident and correct.

---

### Prediction = 0.50

The model is uncertain:

$$
BCE=-\log(1-0.50)
$$

$$
BCE=-\log(0.50)
$$

$$
BCE\approx0.69
$$

So the loss is moderate.

---

### Prediction = 0.99

The model predicts a 99% probability of class 1, but the actual class is 0.

$$
BCE=-\log(1-0.99)
$$

$$
BCE=-\log(0.01)
$$

$$
BCE\approx4.61
$$

The loss is very large.

The model is highly confident but wrong.

---

## Summary

When the actual class is $0$:

| Predicted Probability of Class 1 | Interpretation | BCE Loss |
|---:|---|---:|
| 0.01 | Confident and correct | Very small |
| 0.50 | Uncertain | Moderate |
| 0.99 | Confident and wrong | Very large |

Therefore:

$$
\text{Actual}=0,\quad
\hat{y}\rightarrow0
\Rightarrow
\text{Small Loss}
$$

and:

$$
\text{Actual}=0,\quad
\hat{y}\rightarrow1
\Rightarrow
\text{Large Loss}
$$

---

## Complete BCE Logic

The two cases can be summarized as:

### Actual = 1

$$
BCE=-\log(\hat{y})
$$

The model should predict:

$$
\hat{y}\rightarrow1
$$

### Actual = 0

$$
BCE=-\log(1-\hat{y})
$$

The model should predict:

$$
\hat{y}\rightarrow0
$$

### Key Idea

> BCE rewards the model for assigning a high probability to the correct class and strongly penalizes it when it assigns a high probability to the wrong class.

>

>

>
> $$

# 02_Categorical Cross-Entropy (CCE)

## 1. What is Categorical Cross-Entropy?

Categorical Cross-Entropy (CCE) is a loss function used for multi-class classification problems.

In binary classification, we have two classes:

$$
0,\;1
$$

and commonly use:

Sigmoid + Binary Cross-Entropy

For multi-class classification, we may have three or more classes.

For example:

- Class 0 → Low
- Class 1 → Medium
- Class 2 → High

For this type of problem, we commonly use:

Softmax + Categorical Cross-Entropy

---

## 2. Why Do We Need Softmax?

For multi-class classification, the model needs to produce a probability for each class.

Suppose there are three classes:

$$
Class\ 0,\quad Class\ 1,\quad Class\ 2
$$

The neural network produces three raw outputs called logits:

$$
z_1,\;z_2,\;z_3
$$

Softmax converts these values into probabilities:

$$
P_1,\;P_2,\;P_3
$$

The probabilities satisfy:

$$
0\leq P_i\leq1
$$

and:

$$
P_1+P_2+P_3=1
$$

For example:

| Class | Probability |
|---|---:|
| Class 0 | 0.10 |
| Class 1 | 0.70 |
| Class 2 | 0.20 |

The model's predicted class would be:

$$
\boxed{Class\ 1}
$$

because it has the highest probability.

---

## 3. Categorical Cross-Entropy Equation

The general equation is:

$$
CCE =
-\sum_{i=1}^{C}
y_i\log(\hat{y}_i)
$$

Where:

- $C$ = Number of classes
- $y_i$ = Actual label for class $i$
- $\hat{y}_i$ = Predicted probability for class $i$

For multiple observations, we take the average:

$$
CCE =
-\frac{1}{n}
\sum_{j=1}^{n}
\sum_{i=1}^{C}
y_{ji}\log(\hat{y}_{ji})
$$

---

## 4. Why Does CCE Work?

Suppose the actual class is:

$$
Class\ 1
$$

The correct one-hot label is:

$$
[0,\;1,\;0]
$$

Suppose the model predicts:

$$
[0.10,\;0.80,\;0.10]
$$

The CCE becomes:

$$
CCE =
-\left[
0\log(0.10)
+
1\log(0.80)
+
0\log(0.10)
\right]
$$

Therefore:

$$
CCE=-\log(0.80)
$$

The model assigned a high probability to the correct class, so the loss is relatively small.

---

## 5. What If the Model Is Confidently Wrong?

Actual class:

$$
[0,\;1,\;0]
$$

Prediction:

$$
[0.90,\;0.05,\;0.05]
$$

The model assigned only 0.05 probability to the correct class.

Therefore:

$$
CCE=-\log(0.05)
$$

This produces a large loss.

So, just like BCE:

> CCE strongly penalizes confident incorrect predictions.

---

## 6. One-Hot Encoding

For CCE, the actual class is commonly represented using one-hot encoding.

For three classes:

| Class | One-Hot Representation |
|---|---|
| Class 0 | [1, 0, 0] |
| Class 1 | [0, 1, 0] |
| Class 2 | [0, 0, 1] |

Only the position corresponding to the correct class contains 1.

This allows the CCE equation to select the predicted probability of the correct class.

---

$$

For example:

$$
[2.0,\;4.0,\;1.0]
$$

$$

$$

$$
[0.11,\;0.84,\;0.04]

## 8. BCE vs CCE

| Feature | BCE | CCE |
|---|---|---|
| Typical problem | Binary classification | Multi-class classification |
| Number of classes | 2 | 3 or more |
| Output activation | Sigmoid | Softmax |
| Output | One probability | Probability for each class |
| Probabilities | Independent positive-class probability | Probabilities sum to 1 |
| Loss | Binary Cross-Entropy | Categorical Cross-Entropy |

---

## Key Takeaway

> BCE is commonly used when there are two classes, while CCE is commonly used when there are multiple mutually exclusive classes.

The typical combinations are:

$$
\boxed{\text{Sigmoid + BCE}}
$$

for binary classification, and

$$
\boxed{\text{Softmax + CCE}}
$$

for multi-class classification.

# Multi-Class vs Multi-Label Classification

These two problems can look similar, but they are fundamentally different.

---

## 1. Multi-Class Classification

In multi-class classification, each sample belongs to exactly one class.

For example, suppose we classify a molecule into:

- Class 0 → Low toxicity
- Class 1 → Moderate toxicity
- Class 2 → High toxicity

A molecule can belong to only one of these classes.

Therefore:

$$
Class\ 0\ OR\ Class\ 1\ OR\ Class\ 2
$$

The model commonly uses:

$$
\boxed{\text{Softmax + Categorical Cross-Entropy}}
$$

Softmax produces probabilities that add up to 1:

$$
P_0+P_1+P_2=1
$$

Example:

$$
[0.10,\;0.70,\;0.20]
$$

The model predicts Class 1 because it has the highest probability.

---

# 2. Multi-Label Classification

In multi-label classification, one sample can belong to multiple classes at the same time.

For example, a molecule could simultaneously be:

- Toxic
- Mutagenic
- Hepatotoxic

These are not mutually exclusive.

A molecule could therefore have:

$$
[1,\;1,\;0]
$$

meaning:

- Toxic → Yes
- Mutagenic → Yes
- Hepatotoxic → No

---

## 3. Why Can't We Use Softmax?

Softmax forces all class probabilities to add up to 1.

For example:

$$
[0.6,\;0.3,\;0.1]
$$

But in multi-label classification, multiple classes can independently be true.

We may want:

$$
[0.90,\;0.85,\;0.10]
$$

Here:

- Class 0 → 90%
- Class 1 → 85%
- Class 2 → 10%

The first two classes can both have high probabilities.

Softmax is not appropriate because it assumes the classes compete with each other.

---

# 4. What Do We Use Instead?

For multi-label classification, we commonly use:

$$
\boxed{\text{Sigmoid + Binary Cross-Entropy}}
$$

A separate sigmoid output is used for each label.

For example:

$$
[0.90,\;0.85,\;0.10]
$$

Each probability is independent.

Therefore:

Input

  ↓
Neural Network

  ↓
Multiple Output Neurons

  ↓
Sigmoid for each output

  ↓
Independent probabilities

  ↓
Binary Cross-Entropy

---

# 5. Example

Suppose we want to predict three properties:

Toxicity
Mutagenicity
Hepatotoxicity
Actual labels:

$$
[1,\;1,\;0]
$$

The model predicts:

$$
[0.90,\;0.85,\;0.10]
$$

Interpretation:

| Property | Probability |
|---|---:|
| Toxicity | 90% |
| Mutagenicity | 85% |
| Hepatotoxicity | 10% |

Each output is evaluated independently using Binary Cross-Entropy.

---

# 6. Important Comparison

| Problem | Classes | Output Activation | Loss |
|---|---|---|---|
| Binary classification | 2 | Sigmoid | BCE |
| Multi-class classification | One of many | Softmax | CCE |
| Multi-label classification | Multiple can be true | Sigmoid | BCE |

---

## Simple Memory Trick

### Binary

> One decision → Sigmoid + BCE

### Multi-Class

> One class among many → Softmax + CCE

### Multi-Label

> Multiple classes can be true → Sigmoid + BCE

---

## Key Takeaway

> Softmax is used when classes are mutually exclusive and compete with each other. Sigmoid is used when each class/label can be predicted independently.

For example:

$$
\boxed{\text{Multi-Class: Softmax + CCE}}
$$

$$
\boxed{\text{Multi-Label: Sigmoid + BCE}}
$$
