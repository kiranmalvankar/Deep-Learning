|---|---|---|
| Recurrence | Yes | No |
| Cell state | LSTM only | No |
| Main mechanism | Gates | Attention |
| Sequence processing | Sequential | More parallelizable |
| Long-range interactions | Through recurrent state | Direct attention |
| Training parallelism | Limited | Higher |
| Main complexity | Recurrent computation | Attention computation |

# 38. Transformer vs CNN

| Feature | CNN | Transformer |
|---|---|---|
| Main mechanism | Convolution | Attention |
| Local bias | Strong | Less inherently local |
| Long-range relationships | Indirect / layered | Direct attention |
| Typical application | Images | Sequences and many other modalities |
| Parameter sharing | Across spatial locations | Through attention projections |
| Global interactions | Require deeper/repeated operations | Natural through self-attention |

# 49. Advantages of Transformers

Transformers have several important advantages over traditional recurrent architectures.

## 1. Directly Model Long-Range Relationships

Self-attention allows one token to directly interact with another token, even when they are far apart in the sequence.

This makes it easier to model long-range dependencies compared with passing information sequentially through an RNN.

---

## 2. Highly Parallelizable

Unlike RNNs, Transformers do not require processing the sequence one time step at a time during the main training computation.


This allows Transformers to make much better use of modern parallel hardware such as GPUs and TPUs.

---

## 3. Better Handling of Long-Range Dependencies

Because attention can create direct connections between distant positions, Transformers can effectively model relationships across long sequences.

The relationship does not need to pass through every intermediate recurrent state.

---

## 4. Multiple Relationships Can Be Learned

Multi-head attention allows the model to learn different types of relationships simultaneously.

The outputs of the different heads are then combined.

---

## 5. Powerful Representation Learning

Transformers can learn rich contextual representations from large datasets.

These representations can then be used for different downstream tasks.

---

Transformers have proven highly effective when trained with large datasets and substantial computational resources.

# 50. Limitations of Transformers

Transformers are powerful, but they also have important limitations.

## 1. High Computational Cost

Self-attention requires interactions between sequence positions.

For a sequence of length $n$, the attention matrix has approximately:

$$
n\times n
$$

elements.

Therefore, the core self-attention computation has approximately quadratic scaling with sequence length.

This becomes particularly challenging for very long sequences.

## 5. Molecular Representation Limitations

When Transformers operate on SMILES, the model is working with a sequence representation of a molecular graph.

The same molecule can sometimes have multiple valid SMILES representations.

Therefore, SMILES-based Transformer models can be sensitive to the chosen representation.

This is one reason molecular modeling can also benefit from representations that explicitly model molecular graphs.
