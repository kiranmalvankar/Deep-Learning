# 14. Batch Normalization

## 1. What is Batch Normalization?

Batch Normalization (BatchNorm) is a technique used in neural networks to normalize the activations of a layer during training.

It helps keep the values flowing through the network within a more stable range.

In simple words:

> Batch Normalization normalizes the intermediate activations of a neural network and then allows the network to learn the appropriate scale and shift.

The basic idea is:

Batch Normalization was introduced to make neural-network training more stable and efficient.

# 5. Batch Normalization Step-by-Step

Suppose a mini-batch contains activations:

$$
x_1,x_2,x_3,...,x_m
$$

BatchNorm performs several steps.

---

# 6. Step 1 — Calculate the Mean

For a batch of $m$ values:

$$
\mu_B =
\frac{1}{m}
\sum_{i=1}^{m}x_i
$$

Where:

- $\mu_B$ = batch mean
- $m$ = number of samples in the batch
- $x_i$ = activation value

---

# 7. Step 2 — Calculate Variance

The batch variance is:

$$
\sigma_B^2 =
\frac{1}{m}
\sum_{i=1}^{m}
(x_i-\mu_B)^2
$$

Where:

- $\sigma_B^2$ = batch variance
- $\mu_B$ = batch mean

Variance tells us how spread out the activation values are.

---

# 8. Step 3 — Normalize

The activation is normalized using:

$$
\hat{x}_i =
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
$$

Where:

- $\hat{x}_i$ = normalized activation
- $\epsilon$ = small value for numerical stability

After normalization, the values are approximately:

- Mean ≈ 0
- Variance ≈ 1


---

# 9. Step 4 — Scale and Shift

BatchNorm does not simply force every activation to have mean 0 and variance 1.

It introduces two learnable parameters:

$$
\gamma
$$

and

$$
\beta
$$

The final output is:

$$
\boxed{
y_i =
\gamma\hat{x}_i+\beta
}
$$

Where:

- $\gamma$ = learnable scale parameter
- $\beta$ = learnable shift parameter

Therefore, BatchNorm can learn the appropriate distribution for the network.

---

# 10. Complete Batch Normalization Equation

Combining the steps:

$$
\boxed{
y_i =
\gamma
\frac{x_i-\mu_B}
{\sqrt{\sigma_B^2+\epsilon}}
+
\beta
}
$$

This is the main equation to remember.

---

# 11. Why Are γ and β Needed?

This is an important concept.

Suppose BatchNorm always forced:

- Mean = 0
- Variance = 1


The network might lose the ability to choose a different useful scale and center.

Therefore BatchNorm learns:

γ → Scale

β → Shift

# 17. BatchNorm During Training

During training, BatchNorm uses statistics from the current mini-batch.

It also maintains moving estimates of the mean and variance for use during inference.

---

# 18. BatchNorm During Inference

During inference/testing, we generally do not calculate statistics from the current single prediction batch in the same way as training.

Instead, BatchNorm uses the moving averages accumulated during training.

Conceptually:

This is important because predictions should be stable and should not depend on the composition of the inference batch.

---

# 19. Training vs Inference

| Stage | Statistics used |
|---|---|
| Training | Current mini-batch statistics |
| Inference | Moving averages accumulated during training |

---

# 20. Batch Size Matters

Because BatchNorm calculates statistics from a mini-batch, batch size can matter.

For example:

Batch size = 32


gives statistics from 32 samples.

But:

Batch size = 2


may produce noisier estimates.

Therefore, very small batch sizes can make BatchNorm less effective or unstable in some situations.

# 27. BatchNorm vs Dropout

| Feature | BatchNorm | Dropout |
|---|---|---|
| Main purpose | Stabilize activations / optimization | Reduce overfitting |
| Main mechanism | Normalize activations | Randomly deactivate units |
| Training | Uses batch statistics | Randomly drops units |
| Inference | Uses moving statistics | Dropout disabled |
| Learnable parameters | γ and β | No γ/β equivalent |
| Primary role | Optimization/training stability | Regularization |

---

# 28. BatchNorm vs L1/L2

| Feature | L1 | L2 | BatchNorm |
|---|---|---|---|
| Main mechanism | Penalizes absolute weights | Penalizes squared weights | Normalizes activations |
| Acts mainly on | Weights | Weights | Layer activations |
| Main goal | Regularization | Regularization | Stable/easier optimization |
| Learnable parameters | λ controls penalty | λ controls penalty | γ and β |

# 34. Advantages of Batch Normalization

BatchNorm can provide several benefits:

1. More stable activation distributions

2. Can make optimization easier

3. Can allow efficient training

4. Can sometimes support larger learning rates

5. Can improve training stability

6. Can provide some regularizing effect


---

# 35. Limitations of Batch Normalization

BatchNorm also has limitations.

1. Depends on batch statistics

2. Very small batch sizes can be problematic

3. Adds additional computation

4. Training and inference behave differently

5. Not always beneficial for every architecture

6. Other normalization methods may be preferable
   in some architectures
